```python
from IPython.display import HTML, display
import json

# ==============================
# APP DỰ ĐOÁN NGUY CƠ TIỂU ĐƯỜNG
# Chạy trực tiếp trên Google Colab
# Không upload file - Không lưu file
# ==============================

model_data = {
    "features": [
        "Pregnancies",
        "Glucose",
        "BloodPressure",
        "SkinThickness",
        "Insulin",
        "BMI",
        "DiabetesPedigreeFunction",
        "Age"
    ],
    "invalidZeroCols": [
        "Glucose",
        "BloodPressure",
        "SkinThickness",
        "Insulin",
        "BMI"
    ],
    "medians": [
        3.0,
        117.0,
        72.0,
        29.0,
        125.0,
        32.3,
        0.3725,
        29.0
    ],
    "means": [
        3.8450520833333335,
        121.65625,
        72.38671875,
        29.108072916666668,
        140.671875,
        32.45520833333333,
        0.47187630208333325,
        33.240885416666664
    ],
    "scales": [
        3.3673836124089958,
        30.418462775166116,
        12.088763750899638,
        8.785495707586254,
        86.32680240314153,
        6.870699333517456,
        0.3311128160286291,
        11.752572645994181
    ],
    "coefs": [
        0.38340655475584084,
        1.1990557836969986,
        -0.043331609605193,
        0.02810975020054023,
        -0.07249396311619394,
        0.6935462271046522,
        0.23403046749500656,
        0.147218981086854
    ],
    "intercept": -0.8773552431619498,
    "accuracy": 70.13,
    "rows": 768
}

model_json = json.dumps(model_data, ensure_ascii=False)

html_code = """
<div class="diabetes-system">
<style>
    .diabetes-system {
        font-family: 'Segoe UI', Arial, sans-serif;
        min-height: 100vh;
        background:
            radial-gradient(circle at top left, rgba(14,165,233,0.22), transparent 30%),
            radial-gradient(circle at bottom right, rgba(34,197,94,0.20), transparent 32%),
            linear-gradient(135deg, #f8fafc, #e0f2fe);
        color: #0f172a;
        border-radius: 28px;
        overflow: hidden;
    }

    .app-shell {
        max-width: 1180px;
        margin: auto;
        padding: 28px;
    }

    .navbar {
        display: flex;
        justify-content: space-between;
        align-items: center;
        gap: 18px;
        margin-bottom: 24px;
    }

    .brand {
        display: flex;
        align-items: center;
        gap: 14px;
    }

    .logo {
        width: 56px;
        height: 56px;
        border-radius: 18px;
        background: linear-gradient(135deg, #0284c7, #22c55e);
        color: white;
        display: flex;
        align-items: center;
        justify-content: center;
        font-size: 28px;
        box-shadow: 0 14px 34px rgba(2,132,199,0.28);
    }

    .brand h1 {
        margin: 0;
        color: #075985;
        font-size: 28px;
        letter-spacing: -0.5px;
    }

    .brand p {
        margin: 3px 0 0;
        color: #64748b;
        font-size: 14px;
    }

    .nav-steps {
        display: flex;
        gap: 10px;
        flex-wrap: wrap;
    }

    .step-btn {
        border: none;
        padding: 11px 15px;
        border-radius: 999px;
        font-weight: 800;
        background: rgba(255,255,255,0.9);
        color: #475569;
        cursor: pointer;
        box-shadow: 0 8px 22px rgba(15,23,42,0.08);
        border: 1px solid #e2e8f0;
    }

    .step-btn.active {
        background: linear-gradient(135deg, #0284c7, #22c55e);
        color: white;
        border-color: transparent;
    }

    .page {
        display: none;
        animation: fadeIn 0.35s ease;
    }

    .page.active {
        display: block;
    }

    @keyframes fadeIn {
        from {
            opacity: 0;
            transform: translateY(12px);
        }
        to {
            opacity: 1;
            transform: translateY(0);
        }
    }

    .glass-card {
        background: rgba(255,255,255,0.92);
        border: 1px solid rgba(226,232,240,0.9);
        box-shadow: 0 24px 70px rgba(15,23,42,0.12);
        border-radius: 30px;
        padding: 30px;
        backdrop-filter: blur(14px);
    }

    .intro-grid {
        display: grid;
        grid-template-columns: 1.2fr 0.8fr;
        gap: 24px;
        align-items: stretch;
    }

    .intro-title {
        font-size: 46px;
        line-height: 1.1;
        margin: 10px 0 16px;
        color: #0f172a;
        letter-spacing: -1px;
    }

    .intro-subtitle {
        font-size: 17px;
        line-height: 1.75;
        color: #475569;
        max-width: 720px;
    }

    .badge {
        display: inline-flex;
        align-items: center;
        gap: 8px;
        background: #ecfeff;
        color: #0369a1;
        border: 1px solid #bae6fd;
        padding: 9px 14px;
        border-radius: 999px;
        font-weight: 800;
    }

    .feature-list {
        display: grid;
        gap: 14px;
        margin-top: 24px;
    }

    .feature-item {
        display: flex;
        gap: 12px;
        background: #f8fafc;
        border: 1px solid #e2e8f0;
        padding: 15px;
        border-radius: 18px;
        color: #334155;
        line-height: 1.5;
    }

    .feature-icon {
        width: 34px;
        height: 34px;
        border-radius: 12px;
        background: #dbeafe;
        display: flex;
        align-items: center;
        justify-content: center;
        flex-shrink: 0;
    }

    .intro-side {
        display: grid;
        gap: 16px;
    }

    .stat-box {
        background: linear-gradient(180deg, white, #f8fafc);
        border: 1px solid #e2e8f0;
        border-radius: 24px;
        padding: 22px;
        box-shadow: 0 18px 45px rgba(15,23,42,0.08);
    }

    .stat-num {
        font-size: 34px;
        color: #0284c7;
        font-weight: 900;
        margin-bottom: 6px;
    }

    .stat-label {
        color: #64748b;
        font-size: 14px;
    }

    .main-action {
        margin-top: 26px;
        display: flex;
        gap: 12px;
        flex-wrap: wrap;
    }

    .btn {
        border: none;
        border-radius: 16px;
        padding: 14px 20px;
        font-size: 15px;
        font-weight: 900;
        cursor: pointer;
        transition: 0.22s;
    }

    .btn-primary {
        color: white;
        background: linear-gradient(135deg, #0284c7, #22c55e);
        box-shadow: 0 14px 32px rgba(2,132,199,0.25);
    }

    .btn-primary:hover {
        transform: translateY(-2px);
        box-shadow: 0 20px 40px rgba(2,132,199,0.32);
    }

    .btn-secondary {
        background: #e2e8f0;
        color: #334155;
    }

    .predict-grid {
        display: grid;
        grid-template-columns: 1.05fr 0.95fr;
        gap: 24px;
    }

    .section-heading {
        display: flex;
        justify-content: space-between;
        align-items: center;
        gap: 12px;
        margin-bottom: 20px;
    }

    .section-heading h2 {
        margin: 0;
        font-size: 28px;
        color: #0f172a;
    }

    .section-heading p {
        margin: 6px 0 0;
        color: #64748b;
        font-size: 14px;
    }

    .form-grid {
        display: grid;
        grid-template-columns: 1fr 1fr;
        gap: 15px;
    }

    .field {
        padding: 14px;
        border-radius: 18px;
        background: #f8fafc;
        border: 1px solid #e2e8f0;
        transition: 0.2s;
    }

    .field:focus-within {
        background: white;
        border-color: #38bdf8;
        box-shadow: 0 0 0 4px rgba(56,189,248,0.15);
    }

    .field label {
        display: block;
        font-size: 13px;
        font-weight: 900;
        color: #334155;
        margin-bottom: 7px;
    }

    .field input {
        width: 100%;
        border: none;
        outline: none;
        background: transparent;
        font-size: 18px;
        font-weight: 800;
        color: #0f172a;
    }

    .field small {
        display: block;
        margin-top: 6px;
        color: #64748b;
        font-size: 12px;
    }

    .preview-card {
        text-align: center;
        background: linear-gradient(180deg, #ffffff, #f8fafc);
        border: 1px solid #e2e8f0;
        border-radius: 26px;
        padding: 26px;
    }

    .risk-circle {
        width: 210px;
        height: 210px;
        margin: 12px auto 20px;
        border-radius: 50%;
        background: conic-gradient(#22c55e 0deg, #e2e8f0 0deg);
        display: flex;
        align-items: center;
        justify-content: center;
        position: relative;
    }

    .risk-circle::before {
        content: "";
        position: absolute;
        width: 150px;
        height: 150px;
        border-radius: 50%;
        background: white;
        box-shadow: inset 0 0 18px rgba(15,23,42,0.06);
    }

    .risk-number {
        position: relative;
        font-size: 42px;
        font-weight: 950;
        color: #0f172a;
    }

    .risk-title {
        font-size: 26px;
        font-weight: 950;
        margin-bottom: 8px;
    }

    .risk-text {
        color: #475569;
        line-height: 1.65;
        margin: 0 auto;
        max-width: 440px;
    }

    .risk-low {
        color: #16a34a;
    }

    .risk-mid {
        color: #ca8a04;
    }

    .risk-high {
        color: #dc2626;
    }

    .doctor-grid {
        display: grid;
        grid-template-columns: 0.9fr 1.1fr;
        gap: 24px;
    }

    .doctor-summary {
        background: linear-gradient(180deg, white, #f8fafc);
        border: 1px solid #e2e8f0;
        border-radius: 26px;
        padding: 24px;
        text-align: center;
    }

    .doctor-avatar {
        width: 86px;
        height: 86px;
        margin: 0 auto 14px;
        border-radius: 26px;
        background: linear-gradient(135deg, #0284c7, #22c55e);
        display: flex;
        align-items: center;
        justify-content: center;
        color: white;
        font-size: 42px;
        box-shadow: 0 16px 36px rgba(2,132,199,0.25);
    }

    .advice-list {
        display: grid;
        gap: 13px;
    }

    .advice-card {
        background: #f8fafc;
        border: 1px solid #e2e8f0;
        border-radius: 18px;
        padding: 16px;
        line-height: 1.55;
    }

    .advice-card strong {
        color: #0f172a;
    }

    .advice-card.warning {
        background: #fff7ed;
        border-color: #fed7aa;
        color: #7c2d12;
    }

    .advice-card.good {
        background: #f0fdf4;
        border-color: #bbf7d0;
        color: #14532d;
    }

    .factor-table {
        width: 100%;
        border-collapse: collapse;
        overflow: hidden;
        border-radius: 18px;
        margin-top: 18px;
        font-size: 14px;
    }

    .factor-table th,
    .factor-table td {
        padding: 13px;
        border-bottom: 1px solid #e2e8f0;
        text-align: left;
    }

    .factor-table th {
        background: #f1f5f9;
        color: #334155;
    }

    .factor-table td {
        background: white;
        color: #475569;
    }

    .tag {
        display: inline-block;
        padding: 5px 9px;
        border-radius: 999px;
        font-size: 12px;
        font-weight: 900;
    }

    .tag-red {
        background: #fee2e2;
        color: #991b1b;
    }

    .tag-yellow {
        background: #fef3c7;
        color: #92400e;
    }

    .tag-green {
        background: #dcfce7;
        color: #166534;
    }

    .medical-note {
        margin-top: 22px;
        padding: 16px;
        border-radius: 20px;
        background: #eff6ff;
        border: 1px solid #bfdbfe;
        color: #1e3a8a;
        line-height: 1.6;
        font-size: 14px;
    }

    @media (max-width: 960px) {
        .intro-grid,
        .predict-grid,
        .doctor-grid {
            grid-template-columns: 1fr;
        }

        .navbar {
            flex-direction: column;
            align-items: flex-start;
        }
    }

    @media (max-width: 650px) {
        .app-shell {
            padding: 18px;
        }

        .form-grid {
            grid-template-columns: 1fr;
        }

        .intro-title {
            font-size: 32px;
        }

        .section-heading {
            flex-direction: column;
            align-items: flex-start;
        }
    }
</style>

<div class="app-shell">

    <div class="navbar">
        <div class="brand">
            <div class="logo">🩺</div>
            <div>
                <h1>DiaCare</h1>
                <p>Ứng dụng hỗ trợ đánh giá nguy cơ tiểu đường</p>
            </div>
        </div>

        <div class="nav-steps">
            <button class="step-btn active" id="navIntro" onclick="goPage('intro')">1. Giới thiệu</button>
            <button class="step-btn" id="navPredict" onclick="goPage('predict')">2. Dự đoán</button>
            <button class="step-btn" id="navAdvice" onclick="goPage('advice')">3. Lời khuyên</button>
        </div>
    </div>

    <!-- TRANG 1 -->
    <section class="page active" id="pageIntro">
        <div class="intro-grid">
            <div class="glass-card">
                <div class="badge">Ứng dụng y tế học tập</div>

                <h2 class="intro-title">
                    Theo dõi nguy cơ tiểu đường từ các chỉ số sức khỏe cơ bản
                </h2>

                <p class="intro-subtitle">
                    DiaCare giúp người dùng nhập các chỉ số như glucose, huyết áp, insulin, BMI,
                    tuổi và yếu tố di truyền để ước lượng nguy cơ tiểu đường. Sau khi dự đoán,
                    ứng dụng sẽ đưa ra phần đánh giá và gợi ý điều chỉnh lối sống theo từng yếu tố.
                </p>

                <div class="feature-list">
                    <div class="feature-item">
                        <div class="feature-icon">📊</div>
                        <div>
                            <strong>Dự đoán trực quan</strong><br>
                            Kết quả hiển thị bằng phần trăm nguy cơ và mức đánh giá thấp, trung bình hoặc cao.
                        </div>
                    </div>

                    <div class="feature-item">
                        <div class="feature-icon">🧾</div>
                        <div>
                            <strong>Phân tích từng chỉ số</strong><br>
                            App chỉ ra yếu tố nào nên chú ý như glucose, BMI, huyết áp, insulin hoặc tuổi.
                        </div>
                    </div>

                    <div class="feature-item">
                        <div class="feature-icon">👨‍⚕️</div>
                        <div>
                            <strong>Lời khuyên tham khảo</strong><br>
                            Trang cuối mô phỏng phần tư vấn sức khỏe: nên giảm gì, tăng gì và cần theo dõi gì.
                        </div>
                    </div>
                </div>

                <div class="main-action">
                    <button class="btn btn-primary" onclick="goPage('predict')">
                        Bắt đầu dự đoán
                    </button>

                    <button class="btn btn-secondary" onclick="fillDemoAndGo()">
                        Dùng dữ liệu mẫu
                    </button>
                </div>
            </div>

            <div class="intro-side">
                <div class="stat-box">
                    <div class="stat-num">8</div>
                    <div class="stat-label">Thông số sức khỏe đầu vào</div>
                </div>

                <div class="stat-box">
                    <div class="stat-num" id="dataRows">768</div>
                    <div class="stat-label">Dòng dữ liệu dùng để xây dựng mô hình</div>
                </div>

                <div class="stat-box">
                    <div class="stat-num" id="modelAcc">70.13%</div>
                    <div class="stat-label">Độ chính xác kiểm thử tham khảo</div>
                </div>

                <div class="stat-box">
                    <div class="stat-num">3 lớp</div>
                    <div class="stat-label">Giới thiệu → Dự đoán → Lời khuyên</div>
                </div>
            </div>
        </div>
    </section>

    <!-- TRANG 2 -->
    <section class="page" id="pagePredict">
        <div class="predict-grid">
            <div class="glass-card">
                <div class="section-heading">
                    <div>
                        <h2>Nhập thông số dự đoán</h2>
                        <p>Điền các chỉ số sức khỏe bên dưới rồi bấm dự đoán.</p>
                    </div>
                </div>

                <div class="form-grid">
                    <div class="field">
                        <label>Số lần mang thai</label>
                        <input type="number" id="Pregnancies" value="2" min="0">
                        <small>Pregnancies</small>
                    </div>

                    <div class="field">
                        <label>Glucose</label>
                        <input type="number" id="Glucose" value="120" min="0">
                        <small>Chỉ số đường huyết</small>
                    </div>

                    <div class="field">
                        <label>Huyết áp</label>
                        <input type="number" id="BloodPressure" value="70" min="0">
                        <small>Blood Pressure</small>
                    </div>

                    <div class="field">
                        <label>Độ dày da</label>
                        <input type="number" id="SkinThickness" value="25" min="0">
                        <small>Skin Thickness</small>
                    </div>

                    <div class="field">
                        <label>Insulin</label>
                        <input type="number" id="Insulin" value="80" min="0">
                        <small>Chỉ số insulin</small>
                    </div>

                    <div class="field">
                        <label>BMI</label>
                        <input type="number" id="BMI" value="30" min="0" step="0.1">
                        <small>Chỉ số khối cơ thể</small>
                    </div>

                    <div class="field">
                        <label>Yếu tố di truyền</label>
                        <input type="number" id="DiabetesPedigreeFunction" value="0.45" min="0" step="0.001">
                        <small>Diabetes Pedigree Function</small>
                    </div>

                    <div class="field">
                        <label>Tuổi</label>
                        <input type="number" id="Age" value="30" min="1">
                        <small>Age</small>
                    </div>
                </div>

                <div class="main-action">
                    <button class="btn btn-primary" onclick="predictDiabetes()">
                        Dự đoán ngay
                    </button>

                    <button class="btn btn-secondary" onclick="useLowRiskSample()">
                        Mẫu nguy cơ thấp
                    </button>

                    <button class="btn btn-secondary" onclick="useHighRiskSample()">
                        Mẫu nguy cơ cao
                    </button>

                    <button class="btn btn-secondary" onclick="resetForm()">
                        Làm mới
                    </button>
                </div>

                <div class="medical-note">
                    Những chỉ số bằng 0 ở các mục như glucose, huyết áp, insulin, BMI sẽ được xử lý
                    như dữ liệu thiếu và thay bằng giá trị trung vị của bộ dữ liệu.
                </div>
            </div>

            <div class="glass-card">
                <div class="section-heading">
                    <div>
                        <h2>Kết quả dự đoán</h2>
                        <p>Kết quả sẽ cập nhật sau khi bấm nút dự đoán.</p>
                    </div>
                </div>

                <div class="preview-card">
                    <div class="risk-circle" id="riskCircle">
                        <div class="risk-number" id="riskNumber">--%</div>
                    </div>

                    <div class="risk-title" id="riskTitle">Chưa có kết quả</div>

                    <p class="risk-text" id="riskText">
                        Sau khi dự đoán, bạn có thể chuyển sang trang lời khuyên để xem phân tích chi tiết.
                    </p>

                    <div class="main-action" style="justify-content:center;">
                        <button class="btn btn-primary" onclick="goPage('advice')">
                            Xem đánh giá & lời khuyên
                        </button>
                    </div>
                </div>
            </div>
        </div>
    </section>

    <!-- TRANG 3 -->
    <section class="page" id="pageAdvice">
        <div class="doctor-grid">
            <div class="glass-card">
                <div class="doctor-summary">
                    <div class="doctor-avatar">👨‍⚕️</div>

                    <h2 style="margin:0 0 8px;">
                        Đánh giá sức khỏe
                    </h2>

                    <p style="color:#64748b; line-height:1.6;">
                        Phần này mô phỏng lời khuyên tham khảo dựa trên các chỉ số bạn nhập
                        và kết quả dự đoán.
                    </p>

                    <div class="risk-circle" id="doctorCircle">
                        <div class="risk-number" id="doctorNumber">--%</div>
                    </div>

                    <div class="risk-title" id="doctorTitle">Chưa có kết quả</div>

                    <p class="risk-text" id="doctorText">
                        Hãy quay lại trang dự đoán và nhập thông số trước.
                    </p>
                </div>

                <div class="medical-note">
                    <b>Lưu ý:</b> Đây là ứng dụng học tập, không phải công cụ chẩn đoán y khoa.
                    Nếu nguy cơ cao hoặc có triệu chứng bất thường, nên kiểm tra y tế chính thức.
                </div>
            </div>

            <div class="glass-card">
                <div class="section-heading">
                    <div>
                        <h2>Lời khuyên cá nhân hóa</h2>
                        <p>Ứng dụng sẽ gợi ý nên giảm, tăng hoặc theo dõi yếu tố nào.</p>
                    </div>
                </div>

                <div class="advice-list" id="adviceList">
                    <div class="advice-card">
                        Chưa có dữ liệu. Hãy sang trang dự đoán và bấm <strong>Dự đoán ngay</strong>.
                    </div>
                </div>

                <table class="factor-table">
                    <thead>
                        <tr>
                            <th>Yếu tố</th>
                            <th>Giá trị</th>
                            <th>Đánh giá</th>
                        </tr>
                    </thead>
                    <tbody id="factorTable">
                        <tr>
                            <td colspan="3">Chưa có dữ liệu đánh giá.</td>
                        </tr>
                    </tbody>
                </table>

                <div class="main-action">
                    <button class="btn btn-secondary" onclick="goPage('predict')">
                        Quay lại chỉnh thông số
                    </button>

                    <button class="btn btn-primary" onclick="predictDiabetes(); goPage('advice');">
                        Cập nhật lời khuyên
                    </button>
                </div>
            </div>
        </div>
    </section>

</div>

<script>
    const model = MODEL_JSON_PLACEHOLDER;

    let lastPrediction = null;
    let lastInputs = null;

    document.getElementById("dataRows").innerText = model.rows;
    document.getElementById("modelAcc").innerText = model.accuracy + "%";

    function goPage(page) {
        const pages = ["Intro", "Predict", "Advice"];

        pages.forEach(p => {
            document.getElementById("page" + p).classList.remove("active");
        });

        document.getElementById("navIntro").classList.remove("active");
        document.getElementById("navPredict").classList.remove("active");
        document.getElementById("navAdvice").classList.remove("active");

        if (page === "intro") {
            document.getElementById("pageIntro").classList.add("active");
            document.getElementById("navIntro").classList.add("active");
        }

        if (page === "predict") {
            document.getElementById("pagePredict").classList.add("active");
            document.getElementById("navPredict").classList.add("active");
        }

        if (page === "advice") {
            document.getElementById("pageAdvice").classList.add("active");
            document.getElementById("navAdvice").classList.add("active");
            updateAdvicePage();
        }
    }

    function sigmoid(z) {
        return 1 / (1 + Math.exp(-z));
    }

    function getValue(id) {
        let value = parseFloat(document.getElementById(id).value);
        if (isNaN(value)) return 0;
        return value;
    }

    function getInputs() {
        return {
            Pregnancies: getValue("Pregnancies"),
            Glucose: getValue("Glucose"),
            BloodPressure: getValue("BloodPressure"),
            SkinThickness: getValue("SkinThickness"),
            Insulin: getValue("Insulin"),
            BMI: getValue("BMI"),
            DiabetesPedigreeFunction: getValue("DiabetesPedigreeFunction"),
            Age: getValue("Age")
        };
    }

    function predictDiabetes() {
        let rawInputs = getInputs();
        let values = [];

        for (let i = 0; i < model.features.length; i++) {
            let feature = model.features[i];
            let value = rawInputs[feature];

            if (model.invalidZeroCols.includes(feature) && value === 0) {
                value = model.medians[i];
            }

            values.push(value);
        }

        let scaled = values.map((value, i) => {
            return (value - model.means[i]) / model.scales[i];
        });

        let z = model.intercept;

        for (let i = 0; i < scaled.length; i++) {
            z += scaled[i] * model.coefs[i];
        }

        let probability = sigmoid(z);
        let percent = probability * 100;

        lastPrediction = percent;
        lastInputs = rawInputs;

        renderPrediction(percent);
        updateAdvicePage();
    }

    function getRiskInfo(percent) {
        if (percent < 35) {
            return {
                label: "Nguy cơ thấp",
                level: "Thấp",
                color: "#22c55e",
                cls: "risk-low",
                text: "Các chỉ số hiện tại cho thấy nguy cơ thấp. Tuy nhiên vẫn nên duy trì chế độ ăn uống và vận động hợp lý."
            };
        }

        if (percent < 65) {
            return {
                label: "Nguy cơ trung bình",
                level: "Trung bình",
                color: "#f59e0b",
                cls: "risk-mid",
                text: "Kết quả cho thấy có một số yếu tố cần chú ý. Nên theo dõi đường huyết, BMI và lối sống thường xuyên hơn."
            };
        }

        return {
            label: "Nguy cơ cao",
            level: "Cao",
            color: "#ef4444",
            cls: "risk-high",
            text: "Kết quả cho thấy nguy cơ cao. Nên kiểm tra sức khỏe chính thức hoặc tham khảo ý kiến bác sĩ."
        };
    }

    function renderPrediction(percent) {
        const info = getRiskInfo(percent);
        const deg = Math.min(percent * 3.6, 360);

        const riskCircle = document.getElementById("riskCircle");
        const riskNumber = document.getElementById("riskNumber");
        const riskTitle = document.getElementById("riskTitle");
        const riskText = document.getElementById("riskText");

        riskCircle.style.background = `conic-gradient(${info.color} ${deg}deg, #e2e8f0 ${deg}deg)`;
        riskNumber.innerText = percent.toFixed(1) + "%";
        riskTitle.innerText = info.label;
        riskTitle.className = "risk-title " + info.cls;
        riskText.innerText = info.text;
    }

    function updateAdvicePage() {
        const doctorCircle = document.getElementById("doctorCircle");
        const doctorNumber = document.getElementById("doctorNumber");
        const doctorTitle = document.getElementById("doctorTitle");
        const doctorText = document.getElementById("doctorText");
        const adviceList = document.getElementById("adviceList");
        const factorTable = document.getElementById("factorTable");

        if (lastPrediction === null || lastInputs === null) {
            doctorCircle.style.background = "conic-gradient(#22c55e 0deg, #e2e8f0 0deg)";
            doctorNumber.innerText = "--%";
            doctorTitle.innerText = "Chưa có kết quả";
            doctorTitle.className = "risk-title";
            doctorText.innerText = "Hãy quay lại trang dự đoán và nhập thông số trước.";

            adviceList.innerHTML = `
                <div class="advice-card">
                    Chưa có dữ liệu. Hãy sang trang dự đoán và bấm <strong>Dự đoán ngay</strong>.
                </div>
            `;

            factorTable.innerHTML = `
                <tr>
                    <td colspan="3">Chưa có dữ liệu đánh giá.</td>
                </tr>
            `;

            return;
        }

        const percent = lastPrediction;
        const info = getRiskInfo(percent);
        const deg = Math.min(percent * 3.6, 360);

        doctorCircle.style.background = `conic-gradient(${info.color} ${deg}deg, #e2e8f0 ${deg}deg)`;
        doctorNumber.innerText = percent.toFixed(1) + "%";
        doctorTitle.innerText = info.label;
        doctorTitle.className = "risk-title " + info.cls;
        doctorText.innerText = info.text;

        const advice = buildAdvice(lastInputs, percent);
        adviceList.innerHTML = advice.cards;
        factorTable.innerHTML = advice.rows;
    }

    function tag(text, type) {
        return `<span class="tag tag-${type}">${text}</span>`;
    }

    function buildAdvice(x, percent) {
        let cards = [];
        let rows = [];

        function addRow(name, value, status, type) {
            rows.push(`
                <tr>
                    <td>${name}</td>
                    <td>${value}</td>
                    <td>${tag(status, type)}</td>
                </tr>
            `);
        }

        if (percent >= 65) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Ưu tiên:</strong> Kết quả đang ở mức nguy cơ cao. Bạn nên kiểm tra đường huyết,
                    HbA1c hoặc tham khảo bác sĩ để được đánh giá chính xác hơn.
                </div>
            `);
        } else if (percent >= 35) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Ưu tiên:</strong> Kết quả ở mức trung bình. Nên theo dõi các chỉ số có xu hướng cao
                    và điều chỉnh lối sống sớm.
                </div>
            `);
        } else {
            cards.push(`
                <div class="advice-card good">
                    <strong>Tình trạng hiện tại:</strong> Nguy cơ thấp. Vẫn nên duy trì ăn uống lành mạnh,
                    ngủ đủ, vận động đều và kiểm tra sức khỏe định kỳ.
                </div>
            `);
        }

        if (x.Glucose >= 140) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Glucose đang cao:</strong> Nên giảm đồ uống ngọt, bánh kẹo,
                    tinh bột hấp thu nhanh và theo dõi đường huyết thường xuyên hơn.
                </div>
            `);
            addRow("Glucose", x.Glucose, "Cần chú ý", "red");
        } else if (x.Glucose >= 110) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Glucose hơi cao:</strong> Nên kiểm soát khẩu phần cơm, bún, mì,
                    nước ngọt và tăng rau xanh trong bữa ăn.
                </div>
            `);
            addRow("Glucose", x.Glucose, "Theo dõi", "yellow");
        } else {
            addRow("Glucose", x.Glucose, "Ổn", "green");
        }

        if (x.BMI >= 30) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>BMI cao:</strong> Nên ưu tiên giảm cân từ từ bằng cách giảm năng lượng dư thừa,
                    hạn chế đồ chiên, nước ngọt và tăng vận động.
                </div>
            `);
            addRow("BMI", x.BMI, "Cao", "red");
        } else if (x.BMI >= 25) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>BMI hơi cao:</strong> Nên kiểm soát cân nặng, tăng vận động
                    và theo dõi vòng eo.
                </div>
            `);
            addRow("BMI", x.BMI, "Theo dõi", "yellow");
        } else if (x.BMI < 18.5) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>BMI thấp:</strong> Không nên tự giảm cân. Nên ăn đủ chất
                    và cân nhắc tư vấn dinh dưỡng.
                </div>
            `);
            addRow("BMI", x.BMI, "Thấp", "yellow");
        } else {
            addRow("BMI", x.BMI, "Ổn", "green");
        }

        if (x.BloodPressure >= 140) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Huyết áp cao:</strong> Nên giảm muối, hạn chế thức ăn chế biến sẵn,
                    ngủ đủ và kiểm tra huyết áp định kỳ.
                </div>
            `);
            addRow("Huyết áp", x.BloodPressure, "Cao", "red");
        } else if (x.BloodPressure >= 120) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Huyết áp cần theo dõi:</strong> Nên giảm muối, giảm stress
                    và duy trì vận động nhẹ.
                </div>
            `);
            addRow("Huyết áp", x.BloodPressure, "Theo dõi", "yellow");
        } else {
            addRow("Huyết áp", x.BloodPressure, "Ổn", "green");
        }

        if (x.Insulin >= 180) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Insulin cao:</strong> Có thể liên quan đến tình trạng đề kháng insulin.
                    Nên kiểm soát cân nặng, giảm đường đơn và vận động đều.
                </div>
            `);
            addRow("Insulin", x.Insulin, "Cao", "red");
        } else if (x.Insulin == 0) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Insulin bằng 0:</strong> Giá trị này có thể là dữ liệu thiếu.
                    Khi dùng thực tế nên nhập kết quả xét nghiệm chính xác nếu có.
                </div>
            `);
            addRow("Insulin", x.Insulin, "Thiếu dữ liệu", "yellow");
        } else {
            addRow("Insulin", x.Insulin, "Tham khảo", "green");
        }

        if (x.DiabetesPedigreeFunction >= 0.8) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Yếu tố di truyền cao:</strong> Không thể thay đổi di truyền,
                    nhưng có thể giảm rủi ro bằng kiểm soát cân nặng, ăn uống và vận động.
                </div>
            `);
            addRow("Yếu tố di truyền", x.DiabetesPedigreeFunction, "Cao", "red");
        } else if (x.DiabetesPedigreeFunction >= 0.5) {
            addRow("Yếu tố di truyền", x.DiabetesPedigreeFunction, "Theo dõi", "yellow");
        } else {
            addRow("Yếu tố di truyền", x.DiabetesPedigreeFunction, "Thấp", "green");
        }

        if (x.Age >= 45) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Tuổi là yếu tố nguy cơ:</strong> Khi tuổi tăng, nên kiểm tra sức khỏe định kỳ
                    và theo dõi đường huyết thường xuyên hơn.
                </div>
            `);
            addRow("Tuổi", x.Age, "Cần theo dõi", "yellow");
        } else {
            addRow("Tuổi", x.Age, "Ổn", "green");
        }

        if (x.Pregnancies >= 5) {
            cards.push(`
                <div class="advice-card warning">
                    <strong>Số lần mang thai cao:</strong> Đây là yếu tố nên được theo dõi thêm,
                    đặc biệt nếu từng có tiền sử tiểu đường thai kỳ.
                </div>
            `);
            addRow("Số lần mang thai", x.Pregnancies, "Theo dõi", "yellow");
        } else {
            addRow("Số lần mang thai", x.Pregnancies, "Tham khảo", "green");
        }

        cards.push(`
            <div class="advice-card good">
                <strong>Nên tăng:</strong> vận động đều, rau xanh, chất xơ, uống đủ nước,
                ngủ đủ giấc và kiểm tra sức khỏe định kỳ.
            </div>
        `);

        cards.push(`
            <div class="advice-card warning">
                <strong>Nên giảm:</strong> nước ngọt, bánh kẹo, tinh bột hấp thu nhanh,
                đồ chiên nhiều dầu, ăn khuya và lối sống ít vận động.
            </div>
        `);

        return {
            cards: cards.join(""),
            rows: rows.join("")
        };
    }

    function setValues(data) {
        for (const key in data) {
            document.getElementById(key).value = data[key];
        }
    }

    function useLowRiskSample() {
        setValues({
            Pregnancies: 1,
            Glucose: 90,
            BloodPressure: 66,
            SkinThickness: 23,
            Insulin: 70,
            BMI: 23.5,
            DiabetesPedigreeFunction: 0.25,
            Age: 25
        });

        predictDiabetes();
    }

    function useHighRiskSample() {
        setValues({
            Pregnancies: 6,
            Glucose: 165,
            BloodPressure: 86,
            SkinThickness: 35,
            Insulin: 190,
            BMI: 36.5,
            DiabetesPedigreeFunction: 0.82,
            Age: 52
        });

        predictDiabetes();
    }

    function fillDemoAndGo() {
        useHighRiskSample();
        goPage("predict");
    }

    function resetForm() {
        setValues({
            Pregnancies: 2,
            Glucose: 120,
            BloodPressure: 70,
            SkinThickness: 25,
            Insulin: 80,
            BMI: 30,
            DiabetesPedigreeFunction: 0.45,
            Age: 30
        });

        lastPrediction = null;
        lastInputs = null;

        document.getElementById("riskCircle").style.background =
            "conic-gradient(#22c55e 0deg, #e2e8f0 0deg)";
        document.getElementById("riskNumber").innerText = "--%";
        document.getElementById("riskTitle").innerText = "Chưa có kết quả";
        document.getElementById("riskTitle").className = "risk-title";
        document.getElementById("riskText").innerText =
            "Sau khi dự đoán, bạn có thể chuyển sang trang lời khuyên để xem phân tích chi tiết.";

        updateAdvicePage();
    }
</script>
</div>
"""

html_code = html_code.replace("MODEL_JSON_PLACEHOLDER", model_json)

display(HTML(html_code))
```
