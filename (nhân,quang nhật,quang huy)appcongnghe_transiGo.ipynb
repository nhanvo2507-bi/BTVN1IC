```python
from IPython.display import HTML, display
from pathlib import Path

html_code = r'''
<!DOCTYPE html>
<html lang="vi">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>transiGo AI</title>

<link rel="stylesheet" href="https://unpkg.com/leaflet/dist/leaflet.css"/>
<script src="https://unpkg.com/leaflet/dist/leaflet.js"></script>

<style>
:root{
    --blue:#0056B3;
    --cyan:#00A6FF;
    --orange:#FF6B00;
    --green:#22C55E;
    --red:#EF4444;
    --dark:#0F172A;
    --muted:#64748B;
    --line:#E2E8F0;
    --soft:#F8FAFC;
}

*{
    box-sizing:border-box;
}

body{
    margin:0;
    font-family:'Segoe UI',Arial,sans-serif;
    background:linear-gradient(135deg,#DBEAFE,#FFF7ED);
    color:var(--dark);
    overflow-x:hidden;
}

button,input,select,textarea{
    font-family:inherit;
}

button:disabled,
input:disabled,
select:disabled{
    opacity:.6;
    cursor:not-allowed;
}

/* PAGE 1 */
#landing-page{
    min-height:100vh;
    position:relative;
    overflow:hidden;
    display:flex;
    align-items:center;
    justify-content:center;
    padding:28px;
    background:
        radial-gradient(circle at 15% 20%,rgba(0,166,255,.36),transparent 30%),
        radial-gradient(circle at 85% 15%,rgba(255,107,0,.34),transparent 32%),
        radial-gradient(circle at 70% 85%,rgba(34,197,94,.20),transparent 30%),
        linear-gradient(135deg,#061B4E,#0056B3 45%,#FF6B00);
}

.landing-orb{
    position:absolute;
    border-radius:50%;
    opacity:.35;
    animation:float 5s ease-in-out infinite;
}

.orb-1{
    width:180px;
    height:180px;
    background:#38BDF8;
    top:12%;
    left:7%;
}

.orb-2{
    width:240px;
    height:240px;
    background:#FDBA74;
    bottom:8%;
    right:8%;
    animation-delay:1s;
}

.orb-3{
    width:120px;
    height:120px;
    background:#86EFAC;
    bottom:18%;
    left:18%;
    animation-delay:2s;
}

@keyframes float{
    0%,100%{
        transform:translateY(0);
    }

    50%{
        transform:translateY(-22px);
    }
}

.landing-shell{
    width:min(1150px,100%);
    display:grid;
    grid-template-columns:1.08fr .92fr;
    gap:38px;
    align-items:center;
    position:relative;
    z-index:2;
}

.landing-content{
    color:white;
}

.logo-pill{
    display:inline-flex;
    align-items:center;
    gap:9px;
    background:rgba(255,255,255,.16);
    border:1px solid rgba(255,255,255,.28);
    padding:9px 14px;
    border-radius:999px;
    font-weight:900;
    backdrop-filter:blur(12px);
    margin-bottom:18px;
}

.landing-title{
    font-size:clamp(3rem,7vw,6.4rem);
    line-height:.88;
    letter-spacing:-5px;
    font-weight:1000;
    margin:0;
}

.landing-title span{
    color:#FFE0C2;
}

.landing-subtitle{
    margin:22px 0 0 0;
    max-width:620px;
    font-size:1.2rem;
    line-height:1.55;
    font-weight:700;
    opacity:.92;
}

.landing-actions{
    display:flex;
    gap:14px;
    flex-wrap:wrap;
    margin-top:28px;
}

.primary-big-btn{
    border:none;
    padding:17px 26px;
    border-radius:999px;
    background:white;
    color:var(--blue);
    font-weight:1000;
    cursor:pointer;
    box-shadow:0 22px 45px rgba(0,0,0,.28);
    font-size:1rem;
}

.ghost-big-btn{
    border:1px solid rgba(255,255,255,.38);
    padding:17px 22px;
    border-radius:999px;
    background:rgba(255,255,255,.13);
    color:white;
    font-weight:900;
    cursor:pointer;
    backdrop-filter:blur(10px);
}

.landing-badges{
    display:flex;
    gap:10px;
    flex-wrap:wrap;
    margin-top:28px;
}

.landing-badge{
    background:rgba(255,255,255,.14);
    border:1px solid rgba(255,255,255,.25);
    color:white;
    padding:10px 12px;
    border-radius:999px;
    font-weight:900;
    font-size:.86rem;
    backdrop-filter:blur(10px);
}

.phone-mock{
    width:min(380px,100%);
    margin:auto;
    background:rgba(255,255,255,.18);
    border:1px solid rgba(255,255,255,.35);
    border-radius:42px;
    padding:18px;
    box-shadow:0 36px 90px rgba(0,0,0,.35);
    backdrop-filter:blur(18px);
    transform:rotate(3deg);
}

.phone-screen{
    background:#F8FAFC;
    border-radius:32px;
    overflow:hidden;
    min-height:620px;
    position:relative;
}

.phone-map{
    height:260px;
    background:
        linear-gradient(90deg,rgba(0,86,179,.12) 1px,transparent 1px),
        linear-gradient(rgba(0,86,179,.10) 1px,transparent 1px),
        linear-gradient(135deg,#DBEAFE,#FFF7ED);
    background-size:28px 28px;
    position:relative;
}

.map-route{
    position:absolute;
    left:65px;
    top:70px;
    width:220px;
    height:120px;
    border:8px solid #0056B3;
    border-left-color:transparent;
    border-bottom-color:#FF6B00;
    border-radius:70px;
    transform:rotate(-8deg);
}

.map-pin{
    position:absolute;
    width:54px;
    height:54px;
    border-radius:18px;
    display:flex;
    align-items:center;
    justify-content:center;
    color:white;
    font-size:1.5rem;
    box-shadow:0 14px 28px rgba(15,23,42,.24);
}

.pin-a{
    left:48px;
    top:55px;
    background:#0056B3;
}

.pin-b{
    right:58px;
    bottom:48px;
    background:#FF6B00;
}

.phone-card-stack{
    padding:18px;
}

.mini-card{
    background:white;
    border:1px solid #E2E8F0;
    border-radius:22px;
    padding:16px;
    margin-bottom:12px;
    box-shadow:0 10px 24px rgba(15,23,42,.08);
}

.mini-card strong{
    color:#0056B3;
}

.mini-row{
    display:flex;
    justify-content:space-between;
    gap:12px;
    color:#64748B;
    font-weight:800;
    margin-top:6px;
}

.price-mini{
    background:linear-gradient(135deg,#0056B3,#00A6FF);
    color:white;
    border-radius:24px;
    padding:20px;
    font-weight:1000;
    font-size:1.7rem;
}

/* PAGE 2 */
#menu-page{
    min-height:100vh;
    display:none;
    padding:30px;
    background:
        radial-gradient(circle at 10% 10%,rgba(0,166,255,.20),transparent 28%),
        radial-gradient(circle at 90% 18%,rgba(255,107,0,.22),transparent 28%),
        linear-gradient(135deg,#F8FAFC,#EFF6FF,#FFF7ED);
}

.menu-shell{
    max-width:1180px;
    margin:0 auto;
}

.menu-top{
    display:flex;
    justify-content:space-between;
    align-items:center;
    gap:16px;
    margin-bottom:34px;
}

.brand{
    font-size:2.3rem;
    font-weight:1000;
    color:var(--blue);
    letter-spacing:-1px;
}

.brand span{
    color:var(--orange);
}

.online-pill{
    background:white;
    border:1px solid var(--line);
    box-shadow:0 10px 24px rgba(15,23,42,.08);
    border-radius:999px;
    padding:10px 14px;
    color:#166534;
    font-weight:1000;
}

.menu-hero{
    background:rgba(255,255,255,.78);
    border:1px solid rgba(255,255,255,.9);
    border-radius:38px;
    padding:32px;
    box-shadow:0 24px 60px rgba(15,23,42,.12);
    backdrop-filter:blur(14px);
    margin-bottom:24px;
}

.menu-hero h1{
    margin:0;
    font-size:clamp(2rem,4vw,4rem);
    letter-spacing:-2px;
    color:var(--dark);
}

.menu-hero p{
    color:var(--muted);
    font-weight:800;
    font-size:1.05rem;
    line-height:1.5;
    max-width:720px;
}

.service-grid{
    display:grid;
    grid-template-columns:1fr 1fr;
    gap:22px;
}

.service-card{
    min-height:310px;
    border:none;
    border-radius:36px;
    padding:26px;
    cursor:pointer;
    color:white;
    text-align:left;
    position:relative;
    overflow:hidden;
    box-shadow:0 26px 60px rgba(15,23,42,.16);
    transition:.26s;
}

.service-card:hover{
    transform:translateY(-8px) scale(1.01);
}

.service-card::after{
    content:"";
    position:absolute;
    width:230px;
    height:230px;
    right:-70px;
    top:-70px;
    border-radius:50%;
    background:rgba(255,255,255,.16);
}

.service-ride{
    background:linear-gradient(135deg,#0056B3,#00A6FF);
}

.service-food{
    background:linear-gradient(135deg,#FF6B00,#F59E0B);
}

.service-icon{
    width:78px;
    height:78px;
    border-radius:28px;
    background:rgba(255,255,255,.18);
    border:1px solid rgba(255,255,255,.28);
    display:flex;
    align-items:center;
    justify-content:center;
    font-size:2.6rem;
    margin-bottom:20px;
}

.service-card h2{
    margin:0 0 10px 0;
    font-size:2rem;
}

.service-card p{
    margin:0;
    max-width:390px;
    line-height:1.5;
    font-weight:700;
    opacity:.93;
}

.service-button{
    display:inline-flex;
    margin-top:24px;
    background:white;
    color:var(--dark);
    border-radius:999px;
    padding:12px 15px;
    font-weight:1000;
}

.stat-grid{
    display:grid;
    grid-template-columns:repeat(3,1fr);
    gap:16px;
    margin-top:22px;
}

.stat-card{
    background:white;
    border:1px solid var(--line);
    border-radius:24px;
    padding:18px;
    box-shadow:0 10px 28px rgba(15,23,42,.08);
}

.stat-card b{
    color:var(--blue);
    font-size:1.2rem;
}

.stat-card div{
    color:var(--muted);
    font-weight:800;
    margin-top:4px;
    font-size:.88rem;
}

/* APP PAGE */
#app-page{
    display:none;
    min-height:100vh;
    padding:18px;
}

.app-topbar{
    max-width:1480px;
    margin:0 auto 18px auto;
    display:flex;
    justify-content:space-between;
    align-items:center;
    gap:16px;
}

.back-btn{
    border:none;
    background:#E0F2FE;
    color:var(--blue);
    font-weight:1000;
    border-radius:999px;
    padding:10px 16px;
    cursor:pointer;
}

.status-pill{
    background:white;
    border:1px solid var(--line);
    border-radius:999px;
    padding:10px 15px;
    color:var(--blue);
    font-weight:1000;
    box-shadow:0 8px 24px rgba(15,23,42,.08);
}

.app-layout{
    max-width:1480px;
    margin:auto;
    display:grid;
    grid-template-columns:450px 1fr;
    gap:22px;
}

.panel{
    background:rgba(255,255,255,.94);
    border:1px solid rgba(255,255,255,.85);
    border-radius:32px;
    box-shadow:0 20px 50px rgba(15,23,42,.12);
}

.left-panel{
    padding:24px;
    max-height:88vh;
    overflow-y:auto;
}

.map-panel{
    min-height:780px;
    overflow:hidden;
    position:relative;
}

.notice{
    background:#EFF6FF;
    border:1px solid #BFDBFE;
    color:#1E40AF;
    border-radius:16px;
    padding:12px;
    font-size:.82rem;
    line-height:1.5;
    font-weight:700;
    margin-bottom:16px;
}

.section-title{
    margin:0 0 14px 0;
    color:var(--blue);
    font-size:1.15rem;
    font-weight:1000;
}

.group{
    position:relative;
    background:var(--soft);
    border:1px solid var(--line);
    border-radius:18px;
    padding:12px 14px;
    margin-bottom:14px;
}

.group label{
    display:block;
    font-size:.72rem;
    text-transform:uppercase;
    font-weight:1000;
    color:var(--muted);
    margin-bottom:6px;
    letter-spacing:.4px;
}

.group input,.group select{
    width:100%;
    border:none;
    outline:none;
    background:transparent;
    font-size:1rem;
    font-weight:800;
    color:var(--dark);
}

.suggest-box{
    position:absolute;
    top:105%;
    left:0;
    width:100%;
    background:white;
    border:1px solid var(--line);
    border-radius:18px;
    box-shadow:0 18px 40px rgba(0,0,0,.18);
    max-height:260px;
    overflow-y:auto;
    display:none;
    z-index:999999;
}

.suggest-item{
    padding:13px 15px;
    cursor:pointer;
    border-bottom:1px solid #F1F5F9;
}

.suggest-item:hover{
    background:#FFF7ED;
}

.suggest-name{
    font-weight:1000;
}

.suggest-sub{
    color:var(--muted);
    font-size:.78rem;
    margin-top:4px;
    line-height:1.3;
}

.option-grid{
    display:grid;
    grid-template-columns:1fr 1fr;
    gap:12px;
    margin-bottom:16px;
}

.option-card{
    background:white;
    border:2px solid var(--line);
    border-radius:22px;
    padding:14px;
    cursor:pointer;
    transition:.2s;
}

.option-card:hover{
    transform:translateY(-3px);
    box-shadow:0 10px 28px rgba(0,0,0,.1);
}

.option-card.active{
    border-color:var(--orange);
    background:#FFF7ED;
}

.option-top{
    display:flex;
    align-items:center;
    gap:10px;
    margin-bottom:8px;
}

.option-icon{
    width:42px;
    height:42px;
    border-radius:15px;
    background:linear-gradient(135deg,#0056B3,#00A6FF);
    display:flex;
    align-items:center;
    justify-content:center;
    color:white;
    font-size:1.3rem;
}

.option-name{
    font-weight:1000;
}

.option-desc{
    font-size:.75rem;
    color:var(--muted);
    line-height:1.4;
    font-weight:700;
}

.promo-grid{
    display:grid;
    grid-template-columns:repeat(3,1fr);
    gap:10px;
    margin-bottom:14px;
}

.promo-card{
    background:white;
    border:2px solid var(--line);
    border-radius:18px;
    padding:12px 8px;
    cursor:pointer;
    text-align:center;
    transition:.2s;
}

.promo-card:hover{
    transform:translateY(-3px);
}

.promo-card.active{
    border-color:var(--orange);
    background:#FFF7ED;
}

.promo-code{
    font-weight:1000;
    color:var(--orange);
    font-size:.9rem;
}

.promo-desc{
    font-size:.72rem;
    color:var(--muted);
    margin-top:4px;
    line-height:1.35;
    font-weight:700;
}

/* FOOD */
.food-hero{
    background:linear-gradient(135deg,#FF6B00,#F59E0B);
    border-radius:28px;
    padding:20px;
    color:white;
    margin-bottom:18px;
    box-shadow:0 18px 35px rgba(255,107,0,.28);
}

.food-hero-top{
    display:flex;
    align-items:center;
    justify-content:space-between;
    gap:14px;
}

.food-hero h2{
    margin:0;
    font-size:1.45rem;
}

.food-hero p{
    margin:6px 0 0 0;
    opacity:.9;
    font-weight:700;
    font-size:.88rem;
}

.food-hero-badge{
    background:rgba(255,255,255,.2);
    padding:8px 12px;
    border-radius:999px;
    font-weight:1000;
    white-space:nowrap;
}

.food-toolbar{
    display:flex;
    gap:10px;
    overflow-x:auto;
    padding-bottom:8px;
    margin-bottom:14px;
}

.food-chip{
    border:none;
    background:white;
    border:1px solid var(--line);
    color:var(--muted);
    padding:10px 14px;
    border-radius:999px;
    font-weight:1000;
    cursor:pointer;
    white-space:nowrap;
    box-shadow:0 6px 16px rgba(15,23,42,.06);
}

.food-chip.active{
    background:#FFF7ED;
    color:var(--orange);
    border-color:var(--orange);
}

.food-brand-grid{
    display:grid;
    grid-template-columns:1fr 1fr;
    gap:12px;
    margin-bottom:18px;
}

.food-brand-card{
    background:white;
    border:2px solid var(--line);
    border-radius:24px;
    padding:14px;
    cursor:pointer;
    transition:.2s;
    position:relative;
    overflow:hidden;
}

.food-brand-card:hover{
    transform:translateY(-3px);
    box-shadow:0 14px 30px rgba(15,23,42,.12);
}

.food-brand-card.active{
    border-color:var(--orange);
    background:#FFF7ED;
}

.brand-logo{
    width:52px;
    height:52px;
    border-radius:20px;
    background:linear-gradient(135deg,#FF6B00,#F59E0B);
    display:flex;
    align-items:center;
    justify-content:center;
    color:white;
    font-size:1.7rem;
    margin-bottom:10px;
}

.brand-name{
    font-weight:1000;
    color:var(--blue);
    font-size:1rem;
}

.brand-desc{
    font-size:.76rem;
    color:var(--muted);
    margin-top:5px;
    line-height:1.4;
    font-weight:700;
}

.brand-meta{
    display:flex;
    gap:6px;
    flex-wrap:wrap;
    margin-top:10px;
}

.brand-tag{
    background:#F1F5F9;
    color:#475569;
    padding:5px 8px;
    border-radius:999px;
    font-size:.68rem;
    font-weight:1000;
}

.food-list{
    display:grid;
    grid-template-columns:1fr;
    gap:12px;
    margin-bottom:16px;
}

.food-category-title{
    margin:16px 0 2px 0;
    font-weight:1000;
    color:var(--blue);
    background:#EFF6FF;
    border:1px solid #BFDBFE;
    padding:11px 13px;
    border-radius:16px;
}

.food-card{
    display:grid;
    grid-template-columns:86px 1fr;
    gap:12px;
    background:white;
    border:1px solid var(--line);
    border-radius:22px;
    padding:12px;
    box-shadow:0 8px 22px rgba(15,23,42,.06);
}

.food-thumb{
    width:86px;
    height:86px;
    border-radius:20px;
    background:linear-gradient(135deg,#F8FAFC,#FFE7D1);
    display:flex;
    align-items:center;
    justify-content:center;
    font-size:2.3rem;
    border:1px solid #FED7AA;
}

.food-info h4{
    margin:0;
    color:var(--dark);
    font-size:.98rem;
}

.food-info small{
    display:block;
    color:var(--muted);
    font-weight:700;
    margin:5px 0 8px 0;
    line-height:1.35;
}

.food-price-line{
    display:flex;
    align-items:center;
    justify-content:space-between;
    gap:10px;
}

.food-base-price{
    color:var(--orange);
    font-weight:1000;
}

.size-row{
    display:flex;
    gap:6px;
    margin-top:9px;
}

.size-btn{
    flex:1;
    border:none;
    border-radius:12px;
    padding:8px 6px;
    background:#F1F5F9;
    font-weight:1000;
    cursor:pointer;
    font-size:.74rem;
    color:#334155;
}

.size-btn:hover{
    background:var(--orange);
    color:white;
}

.cart-box{
    background:white;
    border:1px solid var(--line);
    border-radius:26px;
    padding:16px;
    margin-bottom:16px;
    box-shadow:0 12px 28px rgba(15,23,42,.08);
}

.cart-header{
    display:flex;
    justify-content:space-between;
    align-items:center;
    margin-bottom:8px;
}

.cart-title{
    font-weight:1000;
    color:var(--blue);
}

.cart-count{
    background:#FFF7ED;
    color:var(--orange);
    font-size:.75rem;
    font-weight:1000;
    padding:6px 10px;
    border-radius:999px;
}

.cart-empty{
    color:var(--muted);
    font-weight:700;
    font-size:.88rem;
    padding:10px 0;
}

.cart-row{
    display:grid;
    grid-template-columns:1fr auto;
    gap:10px;
    border-bottom:1px dashed #CBD5E1;
    padding:10px 0;
    font-size:.86rem;
}

.cart-item-name{
    font-weight:1000;
    color:var(--dark);
}

.cart-item-meta{
    color:var(--muted);
    font-size:.78rem;
    margin-top:3px;
    font-weight:700;
}

.cart-controls{
    display:flex;
    align-items:center;
    gap:8px;
}

.qty-btn{
    width:28px;
    height:28px;
    border:none;
    border-radius:50%;
    color:white;
    font-weight:1000;
    cursor:pointer;
}

.minus{
    background:var(--red);
}

.plus{
    background:var(--green);
}

.cart-total-box{
    background:#F8FAFC;
    border-radius:18px;
    padding:12px;
    margin-top:12px;
}

.cart-total-row{
    display:flex;
    justify-content:space-between;
    font-weight:1000;
    margin:6px 0;
}

.cart-total-row.main{
    color:var(--orange);
    font-size:1.05rem;
}

.result-card{
    background:linear-gradient(135deg,#0056B3,#007BFF);
    color:white;
    padding:24px;
    border-radius:28px;
    margin:18px 0;
    box-shadow:0 18px 40px rgba(0,86,179,.28);
}

.result-card small{
    font-weight:1000;
    opacity:.85;
}

.result-card h1{
    margin:8px 0;
    font-size:2.1rem;
}

.meta-row{
    display:flex;
    gap:8px;
    flex-wrap:wrap;
    margin-top:12px;
}

.meta-pill{
    background:rgba(255,255,255,.18);
    padding:7px 10px;
    border-radius:999px;
    font-size:.78rem;
    font-weight:1000;
}

.info-card{
    background:#F8FAFC;
    border:1px solid var(--line);
    border-radius:22px;
    padding:16px;
    margin-bottom:16px;
}

.break-row{
    display:flex;
    justify-content:space-between;
    gap:12px;
    border-bottom:1px dashed #CBD5E1;
    padding:9px 0;
    font-size:.88rem;
}

.break-row:last-child{
    border-bottom:none;
}

.break-row b{
    color:var(--blue);
    text-align:right;
}

.btn-row{
    display:grid;
    grid-template-columns:1fr 1fr;
    gap:12px;
}

.btn{
    border:none;
    border-radius:20px;
    padding:16px;
    font-weight:1000;
    cursor:pointer;
    transition:.2s;
}

.btn:hover{
    transform:translateY(-2px);
}

.btn-primary{
    background:var(--orange);
    color:white;
}

.btn-secondary{
    background:#E0F2FE;
    color:var(--blue);
}

.btn-danger{
    background:var(--red);
    color:white;
}

/* MAP */
#map{
    width:100%;
    height:100%;
    min-height:780px;
}

.map-top{
    position:absolute;
    top:16px;
    left:16px;
    right:16px;
    z-index:999;
    display:grid;
    grid-template-columns:1.1fr 1fr 1fr;
    gap:12px;
    pointer-events:none;
}

.map-card{
    background:rgba(255,255,255,.96);
    border:1px solid var(--line);
    border-radius:22px;
    padding:14px 16px;
    box-shadow:0 12px 30px rgba(0,0,0,.15);
    pointer-events:auto;
}

.map-card-title{
    color:var(--blue);
    font-weight:1000;
    margin-bottom:4px;
}

.map-card-value{
    font-weight:800;
    color:var(--dark);
    line-height:1.35;
}

.map-tools{
    position:absolute;
    top:135px;
    right:16px;
    z-index:999;
    display:flex;
    flex-direction:column;
    gap:10px;
}

.map-tool-btn{
    border:none;
    background:rgba(255,255,255,.96);
    border:1px solid var(--line);
    color:var(--blue);
    font-weight:1000;
    border-radius:18px;
    padding:11px 14px;
    cursor:pointer;
    box-shadow:0 10px 24px rgba(0,0,0,.16);
    text-align:left;
}

.map-tool-btn.active{
    background:#FFF7ED;
    color:var(--orange);
    border-color:var(--orange);
}

.weather-badge{
    position:absolute;
    left:16px;
    bottom:16px;
    z-index:999;
    background:rgba(255,255,255,.97);
    border:1px solid var(--line);
    border-radius:26px;
    padding:16px;
    min-width:280px;
    box-shadow:0 18px 40px rgba(0,0,0,.18);
}

.weather-main{
    display:flex;
    align-items:center;
    gap:14px;
}

.weather-icon{
    width:58px;
    height:58px;
    border-radius:20px;
    background:linear-gradient(135deg,#38BDF8,#0056B3);
    display:flex;
    align-items:center;
    justify-content:center;
    font-size:2rem;
    color:white;
}

.point-marker{
    min-width:112px;
    transform:translate(-50%,-100%);
    font-family:'Segoe UI',Arial,sans-serif;
}

.point-marker .pin{
    display:flex;
    align-items:center;
    gap:7px;
    color:white;
    padding:8px 10px;
    border-radius:999px;
    font-size:.78rem;
    font-weight:1000;
    box-shadow:0 10px 24px rgba(15,23,42,.28);
    border:2px solid white;
    white-space:nowrap;
}

.point-marker .pin::after{
    content:"";
    position:absolute;
    left:50%;
    bottom:-7px;
    width:13px;
    height:13px;
    transform:translateX(-50%) rotate(45deg);
    border-right:2px solid white;
    border-bottom:2px solid white;
}

.point-marker.pickup .pin,
.point-marker.pickup .pin::after{
    background:#0056B3;
}

.point-marker.dropoff .pin,
.point-marker.dropoff .pin::after{
    background:#EF4444;
}

.point-marker.restaurant .pin,
.point-marker.restaurant .pin::after{
    background:#FF6B00;
}

.point-marker.delivery .pin,
.point-marker.delivery .pin::after{
    background:#22C55E;
}

.moving-vehicle-marker{
    width:48px;
    height:48px;
    border-radius:18px;
    background:white;
    border:3px solid #FF6B00;
    display:flex;
    align-items:center;
    justify-content:center;
    font-size:1.75rem;
    box-shadow:0 12px 30px rgba(15,23,42,.32);
}

.moving-vehicle-marker.ride{
    border-color:#0056B3;
}

.moving-vehicle-marker.food{
    border-color:#FF6B00;
}

.live-card{
    position:absolute;
    right:16px;
    bottom:16px;
    z-index:999;
    width:360px;
    background:rgba(255,255,255,.97);
    border:1px solid var(--line);
    border-radius:28px;
    padding:18px;
    box-shadow:0 18px 40px rgba(0,0,0,.18);
    display:none;
}

.driver-head{
    display:flex;
    align-items:center;
    gap:12px;
}

.avatar{
    width:56px;
    height:56px;
    border-radius:20px;
    background:linear-gradient(135deg,#FF6B00,#F59E0B);
    color:white;
    display:flex;
    align-items:center;
    justify-content:center;
    font-size:1.6rem;
}

.driver-name{
    font-weight:1000;
}

.driver-sub{
    color:var(--muted);
    font-size:.82rem;
    margin-top:3px;
    font-weight:700;
}

.trip-live-box{
    background:linear-gradient(135deg,#0056B3,#007BFF);
    color:white;
    padding:14px;
    border-radius:20px;
    margin-top:14px;
}

.trip-live-box small{
    opacity:.85;
    font-weight:800;
}

.trip-live-box h3{
    margin:4px 0 0 0;
    font-size:1rem;
}

.pulse-dot{
    display:inline-block;
    width:9px;
    height:9px;
    border-radius:50%;
    background:#22C55E;
    margin-right:6px;
    animation:pulse 1s infinite;
}

@keyframes pulse{
    0%{
        transform:scale(1);
        opacity:1;
    }

    70%{
        transform:scale(1.8);
        opacity:.2;
    }

    100%{
        transform:scale(1);
        opacity:1;
    }
}

.trip-progress{
    margin-top:14px;
    background:#F8FAFC;
    border:1px solid var(--line);
    border-radius:20px;
    padding:14px;
}

.progress-line{
    height:7px;
    background:#E2E8F0;
    border-radius:999px;
    overflow:hidden;
    margin:12px 0;
}

.progress-fill{
    height:100%;
    width:0%;
    background:linear-gradient(90deg,#0056B3,#22C55E);
    border-radius:999px;
    transition:.4s;
}

.trip-stage{
    display:flex;
    justify-content:space-between;
    font-size:.72rem;
    color:var(--muted);
    font-weight:800;
}

.trip-stage span.active{
    color:var(--blue);
}

.driver-actions{
    display:grid;
    grid-template-columns:1fr 1fr;
    gap:10px;
    margin-top:12px;
}

.small-btn{
    border:none;
    border-radius:16px;
    padding:11px;
    font-weight:1000;
    cursor:pointer;
}

.call-btn{
    background:#DCFCE7;
    color:#166534;
}

.chat-btn{
    background:#DBEAFE;
    color:#1D4ED8;
}

/* MODALS */
.modal-overlay,
.food-modal-overlay,
.rating-overlay{
    position:fixed;
    inset:0;
    background:rgba(15,23,42,.72);
    z-index:999999;
    display:none;
    align-items:center;
    justify-content:center;
    padding:18px;
}

.modal-card,
.rating-window{
    background:white;
    border-radius:34px;
    padding:30px;
    max-width:520px;
    width:95%;
    text-align:center;
    box-shadow:0 30px 80px rgba(0,0,0,.35);
}

.loader{
    width:62px;
    height:62px;
    border-radius:50%;
    border:6px solid #E2E8F0;
    border-top:6px solid var(--orange);
    margin:10px auto 20px;
    animation:spin 1s linear infinite;
}

@keyframes spin{
    to{
        transform:rotate(360deg);
    }
}

.food-menu-window{
    width:95%;
    max-width:760px;
    max-height:88vh;
    overflow-y:auto;
    background:white;
    border-radius:32px;
    padding:22px;
    box-shadow:0 30px 80px rgba(0,0,0,.35);
}

.food-window-head{
    display:flex;
    justify-content:space-between;
    align-items:center;
    gap:14px;
    margin-bottom:16px;
}

.food-window-title{
    font-size:1.4rem;
    font-weight:1000;
    color:var(--blue);
}

.food-window-sub{
    color:var(--muted);
    font-weight:700;
    font-size:.9rem;
    margin-top:4px;
}

.close-food-btn{
    border:none;
    width:42px;
    height:42px;
    border-radius:50%;
    background:#F1F5F9;
    color:var(--dark);
    font-weight:1000;
    cursor:pointer;
    font-size:1.2rem;
}

.food-modal-list{
    display:grid;
    grid-template-columns:1fr;
    gap:12px;
}

.rating-title{
    color:var(--blue);
    font-weight:1000;
    font-size:1.45rem;
    margin-bottom:6px;
    text-align:left;
}

.rating-sub{
    color:var(--muted);
    font-weight:700;
    margin-bottom:18px;
    text-align:left;
}

.rating-row{
    display:flex;
    justify-content:space-between;
    align-items:center;
    gap:12px;
    padding:12px 0;
    border-bottom:1px dashed #CBD5E1;
}

.rating-label{
    font-weight:1000;
    color:var(--dark);
    text-align:left;
}

.stars{
    display:flex;
    gap:4px;
}

.star{
    cursor:pointer;
    font-size:1.45rem;
    color:#CBD5E1;
    transition:.15s;
}

.star.active{
    color:#F59E0B;
}

.rating-note{
    width:100%;
    margin-top:16px;
    border:1px solid var(--line);
    border-radius:18px;
    padding:12px;
    outline:none;
    resize:none;
    font-weight:700;
}

@media(max-width:1100px){
    .landing-shell,
    .app-layout{
        grid-template-columns:1fr;
    }

    .phone-mock{
        transform:none;
    }

    .map-panel,
    #map{
        min-height:590px;
    }

    .map-top{
        grid-template-columns:1fr;
    }
}

@media(max-width:720px){
    .service-grid,
    .stat-grid{
        grid-template-columns:1fr;
    }

    .menu-top,
    .food-hero-top{
        flex-direction:column;
        align-items:flex-start;
    }

    .option-grid,
    .promo-grid,
    .food-brand-grid,
    .btn-row{
        grid-template-columns:1fr;
    }

    .landing-title{
        letter-spacing:-2px;
    }
}

@media(max-width:560px){
    #app-page,
    #menu-page{
        padding:10px;
    }

    .food-card{
        grid-template-columns:72px 1fr;
    }

    .food-thumb{
        width:72px;
        height:72px;
        font-size:2rem;
    }

    .live-card,
    .weather-badge{
        left:12px;
        right:12px;
        width:auto;
    }

    .weather-badge{
        bottom:220px;
    }

    .map-tools{
        top:250px;
        right:12px;
    }
}
</style>
</head>

<body>

<!-- PAGE 1 -->
<div id="landing-page">
    <div class="landing-orb orb-1"></div>
    <div class="landing-orb orb-2"></div>
    <div class="landing-orb orb-3"></div>

    <div class="landing-shell">
        <div class="landing-content">
            <div class="logo-pill">⚡ Smart Mobility Platform</div>
            <h1 class="landing-title">transi<span>Go</span><br>AI</h1>
            <p class="landing-subtitle">
                Di chuyển thông minh trong một chạm. Đặt xe, đặt đồ ăn, xem bản đồ, định tuyến,
                thời tiết và giá động theo giờ cao điểm.
            </p>

            <div class="landing-actions">
                <button class="primary-big-btn" onclick="showMenu()">BẮT ĐẦU TRẢI NGHIỆM →</button>
                <button class="ghost-big-btn" onclick="showMenu()">Xem dịch vụ</button>
            </div>

            <div class="landing-badges">
                <div class="landing-badge">🚗 AI Pricing</div>
                <div class="landing-badge">🗺️ Live Map</div>
                <div class="landing-badge">🌦️ Weather Based</div>
                <div class="landing-badge">🍔 Smart Delivery</div>
            </div>
        </div>

        <div class="phone-mock">
            <div class="phone-screen">
                <div class="phone-map">
                    <div class="map-route"></div>
                    <div class="map-pin pin-a">📍</div>
                    <div class="map-pin pin-b">🏁</div>
                </div>
                <div class="phone-card-stack">
                    <div class="mini-card">
                        <strong>Điểm đón</strong>
                        <div class="mini-row"><span>Nguyễn Huệ</span><span>1.2 km</span></div>
                    </div>
                    <div class="mini-card">
                        <strong>Thời tiết</strong>
                        <div class="mini-row"><span>Trời đẹp</span><span>x1.00</span></div>
                    </div>
                    <div class="price-mini">42.000 VNĐ</div>
                </div>
            </div>
        </div>
    </div>
</div>

<!-- PAGE 2 -->
<div id="menu-page">
    <div class="menu-shell">
        <div class="menu-top">
            <div class="brand">transi<span>Go</span> AI</div>
            <div class="online-pill">● System Online</div>
        </div>

        <div class="menu-hero">
            <h1>Bạn muốn sử dụng dịch vụ nào?</h1>
            <p>
                Chọn một dịch vụ bên dưới để bắt đầu. App sẽ tự tính tuyến đường, giá tiền,
                thời tiết và mô phỏng tài xế/shipper di chuyển trên bản đồ.
            </p>
        </div>

        <div class="service-grid">
            <button class="service-card service-ride" onclick="openService('ride')">
                <div class="service-icon">🚗</div>
                <h2>Đặt xe</h2>
                <p>Tính giá theo quãng đường, thời tiết, giờ cao điểm và mô phỏng tài xế đến đón/trả khách.</p>
                <div class="service-button">Đặt xe ngay →</div>
            </button>

            <button class="service-card service-food" onclick="openService('food')">
                <div class="service-icon">🍔</div>
                <h2>Đồ ăn & nước uống</h2>
                <p>Chọn thương hiệu, xem thực đơn trong cửa sổ riêng, thêm món và theo dõi shipper giao hàng.</p>
                <div class="service-button">Đặt đồ ăn →</div>
            </button>
        </div>

        <div class="stat-grid">
            <div class="stat-card"><b>24/7</b><div>Hoạt động liên tục</div></div>
            <div class="stat-card"><b>AI Fare</b><div>Giá động theo điều kiện</div></div>
            <div class="stat-card"><b>Live Map</b><div>Bản đồ và định tuyến thật</div></div>
        </div>
    </div>
</div>

<!-- APP PAGE -->
<div id="app-page">
    <div class="app-topbar">
        <div>
            <button class="back-btn" onclick="backToMenu()">← Menu</button>
            <div class="brand" style="font-size:2rem;margin-top:8px;">transi<span>Go</span> AI</div>
            <div id="mode-subtitle" style="color:#64748B;font-weight:800;">Đặt xe thông minh</div>
        </div>
        <div class="status-pill" id="system-status">● Sẵn sàng</div>
    </div>

    <div class="app-layout">
        <div class="panel left-panel">
            <div class="notice">
                Bản này đã bỏ GPS và bỏ vé bus. Người dùng nhập địa điểm, chọn gợi ý hoặc bật nút “Chọn trên bản đồ”.
            </div>

            <div id="ride-section">
                <h2 class="section-title">📍 Vị trí chuyến đi</h2>

                <div class="group">
                    <label>Điểm đón</label>
                    <input id="ride-pickup-input" placeholder="Nhập điểm đón..." autocomplete="off" oninput="searchPlaceDebounced(this,'ridePickup')">
                    <div class="suggest-box"></div>
                </div>

                <div class="group">
                    <label>Điểm trả khách</label>
                    <input id="ride-dest-input" placeholder="Bạn muốn đi đâu?" autocomplete="off" oninput="searchPlaceDebounced(this,'rideDest')">
                    <div class="suggest-box"></div>
                </div>

                <h2 class="section-title">🚘 Loại xe</h2>
                <div class="option-grid" id="vehicle-list"></div>
            </div>

            <div id="food-section" style="display:none;">
                <div class="food-hero">
                    <div class="food-hero-top">
                        <div>
                            <h2>Đặt đồ ăn & nước uống</h2>
                            <p>Chọn thương hiệu yêu thích, xem thực đơn và giao đến tận nơi.</p>
                        </div>
                        <div class="food-hero-badge">⚡ Giao nhanh</div>
                    </div>
                </div>

                <h2 class="section-title">📍 Địa chỉ nhận hàng</h2>
                <div class="group">
                    <label>Điểm nhận hàng</label>
                    <input id="food-drop-input" placeholder="Nhập địa chỉ nhận đồ ăn..." autocomplete="off" oninput="searchPlaceDebounced(this,'foodDrop')">
                    <div class="suggest-box"></div>
                </div>

                <h2 class="section-title">🏪 Thương hiệu nổi bật</h2>
                <div class="food-toolbar">
                    <button class="food-chip active" onclick="filterFoodBrands('all', this)">Tất cả</button>
                    <button class="food-chip" onclick="filterFoodBrands('fastfood', this)">Fast Food</button>
                    <button class="food-chip" onclick="filterFoodBrands('coffee', this)">Cà phê</button>
                    <button class="food-chip" onclick="filterFoodBrands('tea', this)">Trà sữa</button>
                    <button class="food-chip" onclick="filterFoodBrands('pizza', this)">Pizza</button>
                </div>

                <div class="food-brand-grid" id="restaurant-list"></div>

                <h2 class="section-title">🍽️ Thực đơn</h2>
                <div class="food-list" id="food-list"></div>

                <div class="cart-box">
                    <div class="cart-header">
                        <div class="cart-title">🛒 Giỏ hàng</div>
                        <div class="cart-count" id="cart-count">0 món</div>
                    </div>
                    <div id="cart-items">Chưa có món nào.</div>
                </div>
            </div>

            <h2 class="section-title">🎟️ Ưu đãi</h2>
            <div class="promo-grid" id="promo-grid">
                <div class="promo-card" onclick="selectPromo('GO20', this)">
                    <div class="promo-code">GO20</div>
                    <div class="promo-desc">Giảm 20% tối đa 30k</div>
                </div>
                <div class="promo-card" onclick="selectPromo('NEWUSER', this)">
                    <div class="promo-code">NEWUSER</div>
                    <div class="promo-desc">Khách mới giảm tối đa 50k</div>
                </div>
                <div class="promo-card" onclick="selectPromo('STUDENT', this)">
                    <div class="promo-code">STUDENT</div>
                    <div class="promo-desc">Sinh viên giảm tối đa 20k</div>
                </div>
            </div>

            <div class="group">
                <label>💳 Thanh toán</label>
                <select id="payment-select">
                    <option>Tiền mặt</option>
                    <option>Ví điện tử</option>
                    <option>Thẻ ngân hàng</option>
                </select>
            </div>

            <div class="result-card">
                <small id="price-label">DỰ KIẾN THANH TOÁN</small>
                <h1 id="total-price">0 VNĐ</h1>
                <div id="promo-note" style="font-size:.85rem;opacity:.9;"></div>
                <div class="meta-row">
                    <div class="meta-pill" id="distance-meta">0 km</div>
                    <div class="meta-pill" id="duration-meta">0 phút</div>
                    <div class="meta-pill" id="weather-meta">Thời tiết --</div>
                    <div class="meta-pill" id="peak-meta">Cao điểm --</div>
                </div>
            </div>

            <div class="info-card">
                <div class="section-title" style="font-size:1rem;margin-bottom:8px;">📊 Phân tích giá</div>
                <div id="breakdown"></div>
            </div>

            <div class="btn-row">
                <button class="btn btn-secondary" onclick="resetCurrentService()">LÀM MỚI</button>
                <button class="btn btn-primary" id="book-btn" onclick="confirmService()">XÁC NHẬN</button>
            </div>
        </div>

        <div class="panel map-panel">
            <div class="map-top">
                <div class="map-card">
                    <div class="map-card-title">🗺️ Tuyến đường</div>
                    <div class="map-card-value" id="route-status">Chọn vị trí để tính tuyến đường</div>
                </div>

                <div class="map-card">
                    <div class="map-card-title">⏰ Giờ cao điểm</div>
                    <div class="map-card-value" id="peak-map-status">Đang kiểm tra...</div>
                </div>

                <div class="map-card">
                    <div class="map-card-title">🚦 Mật độ di chuyển</div>
                    <div class="map-card-value" id="traffic-map-status">Tự động theo giờ và tuyến</div>
                </div>
            </div>

            <div id="map"></div>

            <div class="map-tools">
                <button class="map-tool-btn" id="map-pick-btn" onclick="toggleMapPickMode()">📍 Chọn trên bản đồ: Tắt</button>
                <button class="map-tool-btn active" id="follow-btn" onclick="toggleFollowMode()">🎯 Tự theo dõi: Bật</button>
                <button class="map-tool-btn" onclick="focusCurrentRoute()">🗺️ Xem lại tuyến đường</button>
            </div>

            <div class="weather-badge">
                <div class="weather-main">
                    <div class="weather-icon" id="weather-icon">🌤️</div>
                    <div>
                        <b id="weather-title" style="color:#0056B3;">Thời tiết ngoài trời</b>
                        <div id="weather-detail" style="color:#64748B;font-weight:800;margin-top:4px;">Chưa có dữ liệu</div>
                    </div>
                </div>
            </div>

            <div class="live-card" id="live-card">
                <div class="driver-head">
                    <div class="avatar" id="live-avatar">🧑‍✈️</div>
                    <div>
                        <div class="driver-name" id="live-name">Nguyễn Minh Hoàng</div>
                        <div class="driver-sub" id="live-vehicle">Honda City • 51A-888.88</div>
                        <div class="driver-sub" id="live-rating">⭐ 4.9 • 1.248 chuyến</div>
                    </div>
                </div>

                <div class="trip-live-box">
                    <small><span class="pulse-dot"></span>TRẠNG THÁI</small>
                    <h3 id="live-trip-status">Đang xử lý đơn</h3>
                </div>

                <div class="trip-progress">
                    <div class="trip-stage">
                        <span id="stage-1" class="active">Nhận đơn</span>
                        <span id="stage-2">Đến điểm đầu</span>
                        <span id="stage-3">Di chuyển</span>
                        <span id="stage-4">Hoàn thành</span>
                    </div>
                    <div class="progress-line">
                        <div class="progress-fill" id="trip-progress-fill"></div>
                    </div>
                </div>

                <div class="info-card" style="margin:14px 0 0 0;">
                    <div class="break-row">
                        <span>Thời gian còn lại</span>
                        <b id="live-eta">-- phút</b>
                    </div>
                    <div class="break-row">
                        <span>Khoảng cách còn lại</span>
                        <b id="live-distance">-- km</b>
                    </div>
                    <div class="break-row">
                        <span>Mã đơn/chuyến</span>
                        <b id="order-code">TG-0000</b>
                    </div>
                </div>

                <div class="driver-actions">
                    <button class="small-btn call-btn" onclick="fakeCall()">📞 Gọi</button>
                    <button class="small-btn chat-btn" onclick="fakeChat()">💬 Nhắn</button>
                </div>

                <button class="btn btn-danger" style="width:100%;margin-top:14px;" onclick="cancelLiveOrder()">HỦY</button>
            </div>
        </div>
    </div>
</div>

<!-- FOOD MODAL -->
<div class="food-modal-overlay" id="food-menu-modal">
    <div class="food-menu-window">
        <div class="food-window-head">
            <div>
                <div class="food-window-title" id="food-modal-title">Thực đơn</div>
                <div class="food-window-sub" id="food-modal-sub">Chọn món bạn muốn thêm vào giỏ hàng</div>
            </div>
            <button class="close-food-btn" onclick="closeFoodMenuModal()">×</button>
        </div>
        <div class="food-modal-list" id="food-modal-list"></div>
    </div>
</div>

<!-- PROCESS MODAL -->
<div class="modal-overlay" id="process-modal">
    <div class="modal-card">
        <div class="loader" id="modal-loader"></div>
        <h2 id="modal-title" style="color:#0056B3;">Đang xử lý...</h2>
        <p id="modal-text" style="color:#64748B;font-weight:700;">Hệ thống đang tìm đối tác gần bạn.</p>
        <button class="btn btn-secondary" id="modal-close" style="display:none;width:100%;" onclick="closeModal()">ĐÓNG</button>
    </div>
</div>

<!-- RATING MODAL -->
<div class="rating-overlay" id="rating-modal">
    <div class="rating-window">
        <div class="rating-title" id="rating-title">Đánh giá dịch vụ</div>
        <div class="rating-sub" id="rating-sub">Chia sẻ trải nghiệm của bạn</div>

        <div id="rating-criteria"></div>

        <textarea class="rating-note" id="rating-note" rows="3" placeholder="Nhận xét thêm nếu có..."></textarea>

        <button class="btn btn-primary rating-submit-btn" style="width:100%;margin-top:16px;" onclick="submitRating()">
            GỬI ĐÁNH GIÁ
        </button>
    </div>
</div>

<script>
let currentMode="ride";
let orderLocked=false;

let map;
let pickupMarker=null;
let destMarker=null;
let routeLayer=null;
let trafficLayer=null;
let peakCircle=null;
let movingMarker=null;
let approachLayer=null;
let liveTimer=null;

let ridePickup=null;
let rideDest=null;
let foodDrop=null;

let routeDistanceKm=0;
let routeDurationMin=0;
let weatherMultiplier=1;
let peakMultiplier=1;
let trafficMultiplier=1;
let finalPrice=0;
let debounceTimer=null;

let mapPickMode=false;
let followMoving=true;

let selectedVehicle="bike";
let selectedRestaurant="kfc";
let currentBrandFilter="all";
let cart=[];
let selectedPromo="";
let ratingData={};

const NOMINATIM="https://nominatim.openstreetmap.org";
const OSRM="https://router.project-osrm.org";
const OPEN_METEO="https://api.open-meteo.com/v1/forecast";

const defaultPoints={
    ridePickup:{lat:10.7757,lng:106.7039,name:"Phố đi bộ Nguyễn Huệ"},
    rideDest:{lat:10.7952,lng:106.7218,name:"Landmark 81"},
    foodDrop:{lat:10.7702,lng:106.6698,name:"Vạn Hạnh Mall"}
};

const vehicles={
    bike:{name:"Bike",icon:"🏍️",desc:"Nhanh, rẻ, phù hợp đi gần",base:9000,perKm:5200,min:12000,platform:3000,speedFactor:1.15},
    car:{name:"Car",icon:"🚗",desc:"Xe 4 chỗ, phổ biến",base:18000,perKm:11500,min:30000,platform:6000,speedFactor:1},
    premium:{name:"Premium",icon:"🚘",desc:"Xe tốt hơn, tài xế chất lượng",base:28000,perKm:15500,min:45000,platform:9000,speedFactor:.98},
    xl:{name:"XL",icon:"🚐",desc:"Xe 7 chỗ, nhóm đông",base:32000,perKm:17500,min:55000,platform:11000,speedFactor:.9}
};

const restaurants={
    kfc:{name:"KFC",icon:"🍗",type:"fastfood",desc:"Gà rán, burger, combo gia đình",lat:10.7757,lng:106.7039,baseShip:14000,perKm:4500,rating:"4.8",time:"20-30 phút"},
    lotteria:{name:"Lotteria",icon:"🍔",type:"fastfood",desc:"Burger, gà rán, khoai tây, nước",lat:10.7721,lng:106.6983,baseShip:13000,perKm:4300,rating:"4.7",time:"20-35 phút"},
    highlands:{name:"Highlands Coffee",icon:"☕",type:"coffee",desc:"Cà phê, trà, freeze, bánh ngọt",lat:10.7798,lng:106.6990,baseShip:12000,perKm:4200,rating:"4.9",time:"15-25 phút"},
    phuclong:{name:"Phúc Long",icon:"🍵",type:"tea",desc:"Trà sữa, trà đào, cà phê, bánh",lat:10.7717,lng:106.7044,baseShip:12000,perKm:4200,rating:"4.8",time:"15-25 phút"},
    pizza4ps:{name:"Pizza 4P's",icon:"🍕",type:"pizza",desc:"Pizza, pasta, salad, món Âu",lat:10.7829,lng:106.6959,baseShip:16000,perKm:5000,rating:"4.9",time:"30-45 phút"},
    starbucks:{name:"Starbucks",icon:"🥤",type:"coffee",desc:"Frappuccino, latte, cold brew, bánh",lat:10.7952,lng:106.7218,baseShip:15000,perKm:4800,rating:"4.8",time:"20-30 phút"}
};

const brandMenus={
    kfc:[
        {category:"Combo nổi bật",items:[
            {name:"Combo Gà Rán 2 Miếng",desc:"2 miếng gà + khoai tây + nước",prices:{S:79000,M:89000,L:99000}},
            {name:"Burger Zinger Combo",desc:"Burger gà cay + khoai + nước",prices:{S:85000,M:95000,L:105000}},
            {name:"Combo Gia Đình",desc:"Gà rán, khoai, nước cho nhóm",prices:{S:159000,M:199000,L:249000}}
        ]},
        {category:"Món lẻ",items:[
            {name:"Gà Rán Truyền Thống",desc:"Gà rán giòn vị truyền thống",prices:{S:39000,M:69000,L:99000}},
            {name:"Khoai Tây Chiên",desc:"Khoai chiên giòn",prices:{S:29000,M:39000,L:49000}},
            {name:"Pepsi",desc:"Nước ngọt có gas",prices:{S:19000,M:25000,L:30000}}
        ]}
    ],
    lotteria:[
        {category:"Burger & Combo",items:[
            {name:"Burger Bulgogi",desc:"Burger bò sốt Bulgogi",prices:{S:59000,M:79000,L:95000}},
            {name:"Burger Gà Cay",desc:"Burger gà cay + rau + sốt",prices:{S:55000,M:75000,L:89000}},
            {name:"Combo Gà Sốt H&S",desc:"Gà sốt cay ngọt + khoai + nước",prices:{S:79000,M:99000,L:119000}}
        ]},
        {category:"Ăn kèm",items:[
            {name:"Khoai Tây Lắc Phô Mai",desc:"Khoai tây chiên phủ phô mai",prices:{S:35000,M:45000,L:55000}},
            {name:"Mực Khoanh Chiên",desc:"Mực chiên giòn",prices:{S:49000,M:69000,L:89000}},
            {name:"Trà Chanh",desc:"Trà chanh mát lạnh",prices:{S:19000,M:25000,L:30000}}
        ]}
    ],
    highlands:[
        {category:"Cà phê",items:[
            {name:"Phin Sữa Đá",desc:"Cà phê phin Việt Nam",prices:{S:39000,M:49000,L:59000}},
            {name:"Bạc Xỉu",desc:"Cà phê sữa vị nhẹ",prices:{S:39000,M:49000,L:59000}},
            {name:"Caramel Macchiato",desc:"Cà phê sữa caramel",prices:{S:59000,M:69000,L:79000}}
        ]},
        {category:"Freeze & Trà",items:[
            {name:"Freeze Trà Xanh",desc:"Đá xay vị trà xanh",prices:{S:59000,M:69000,L:79000}},
            {name:"Freeze Chocolate",desc:"Đá xay chocolate",prices:{S:59000,M:69000,L:79000}},
            {name:"Trà Sen Vàng",desc:"Trà sen, thạch, kem sữa",prices:{S:49000,M:59000,L:69000}}
        ]}
    ],
    phuclong:[
        {category:"Trà đặc trưng",items:[
            {name:"Trà Đào",desc:"Trà đào thơm, topping đào",prices:{S:45000,M:55000,L:65000}},
            {name:"Trà Vải",desc:"Trà vải thanh mát",prices:{S:45000,M:55000,L:65000}},
            {name:"Trà Sữa Phúc Long",desc:"Trà sữa vị truyền thống",prices:{S:49000,M:59000,L:69000}}
        ]},
        {category:"Cà phê & Bánh",items:[
            {name:"Cà Phê Sữa Đá",desc:"Cà phê sữa đậm vị",prices:{S:39000,M:49000,L:59000}},
            {name:"Latte",desc:"Cà phê sữa kiểu Ý",prices:{S:55000,M:65000,L:75000}},
            {name:"Bánh Mousse Chocolate",desc:"Bánh mềm vị chocolate",prices:{S:49000,M:59000,L:69000}}
        ]}
    ],
    pizza4ps:[
        {category:"Pizza",items:[
            {name:"Margherita Pizza",desc:"Pizza phô mai, cà chua, basil",prices:{S:129000,M:189000,L:249000}},
            {name:"Burrata Parma Ham Pizza",desc:"Pizza phô mai Burrata và ham",prices:{S:189000,M:269000,L:349000}},
            {name:"Seafood Pizza",desc:"Pizza hải sản sốt đặc biệt",prices:{S:159000,M:229000,L:299000}}
        ]},
        {category:"Pasta & Salad",items:[
            {name:"Carbonara",desc:"Mì Ý sốt kem trứng",prices:{S:119000,M:159000,L:199000}},
            {name:"Crab Tomato Pasta",desc:"Mì Ý cua sốt cà chua",prices:{S:139000,M:179000,L:229000}},
            {name:"Caesar Salad",desc:"Salad rau, phô mai, sốt Caesar",prices:{S:89000,M:119000,L:149000}}
        ]}
    ],
    starbucks:[
        {category:"Coffee",items:[
            {name:"Caffè Latte",desc:"Espresso và sữa nóng",prices:{S:65000,M:75000,L:85000}},
            {name:"Cold Brew",desc:"Cà phê ủ lạnh",prices:{S:69000,M:79000,L:89000}},
            {name:"Caramel Macchiato",desc:"Cà phê caramel đặc trưng",prices:{S:75000,M:85000,L:95000}}
        ]},
        {category:"Frappuccino & Bakery",items:[
            {name:"Java Chip Frappuccino",desc:"Đá xay chocolate chip",prices:{S:85000,M:95000,L:105000}},
            {name:"Matcha Frappuccino",desc:"Đá xay trà xanh",prices:{S:85000,M:95000,L:105000}},
            {name:"Blueberry Muffin",desc:"Bánh muffin việt quất",prices:{S:59000,M:69000,L:79000}}
        ]}
    ]
};

const foodIcons={
    "gà":"🍗",
    "burger":"🍔",
    "khoai":"🍟",
    "pepsi":"🥤",
    "trà":"🍵",
    "cà phê":"☕",
    "latte":"☕",
    "freeze":"🥤",
    "pizza":"🍕",
    "pasta":"🍝",
    "salad":"🥗",
    "muffin":"🧁",
    "bánh":"🍰",
    "mực":"🦑",
    "combo":"🍱"
};

const drivers=[
    {name:"Nguyễn Minh Hoàng",vehicle:"Honda City",plate:"51A-888.88",icon:"🧑‍✈️"},
    {name:"Trần Quốc Huy",vehicle:"Toyota Vios",plate:"51G-239.11",icon:"🧑‍✈️"},
    {name:"Lê Nhật Nam",vehicle:"VinFast VF5",plate:"50H-774.20",icon:"🧑‍✈️"}
];

const shippers=[
    {name:"Shipper An",vehicle:"Honda Air Blade",plate:"59X3-221.05",icon:"🛵"},
    {name:"Shipper Bình",vehicle:"Yamaha Grande",plate:"59V2-832.10",icon:"🛵"},
    {name:"Shipper Khoa",vehicle:"VinFast Klara",plate:"59E1-120.88",icon:"🛵"}
];

const ratingCriteriaByMode={
    ride:["Thân thiện","An toàn","Đúng giờ","Xe sạch sẽ","Lái xe chuyên nghiệp"],
    food:["Shipper thân thiện","Giao hàng nhanh","Món còn nóng/lạnh đúng yêu cầu","Đóng gói cẩn thận","Chất lượng thương hiệu"]
};

function showMenu(){
    document.getElementById("landing-page").style.display="none";
    document.getElementById("menu-page").style.display="block";
    document.getElementById("app-page").style.display="none";
}

function backToMenu(){
    if(orderLocked){
        alert("Bạn đang có đơn/chuyến đang chạy, không thể quay lại menu lúc này.");
        return;
    }

    document.getElementById("app-page").style.display="none";
    document.getElementById("menu-page").style.display="block";
}

function openService(mode){
    if(orderLocked){
        alert("Bạn đang có đơn/chuyến đang chạy, không thể đổi dịch vụ.");
        return;
    }

    currentMode=mode;

    document.getElementById("landing-page").style.display="none";
    document.getElementById("menu-page").style.display="none";
    document.getElementById("app-page").style.display="block";

    initMap();
    renderModeUI();
    calculatePeakAndTraffic();
    loadDefaultPositionForMode();

    setTimeout(()=>map.invalidateSize(),300);
}

function initMap(){
    if(map) return;

    map=L.map("map").setView([10.7769,106.7009],13);

    L.tileLayer("https://{s}.tile.openstreetmap.org/{z}/{x}/{y}.png",{
        attribution:"© OpenStreetMap"
    }).addTo(map);

    map.on("dragstart",()=>{
        followMoving=false;
        updateFollowButton();
    });

    map.on("zoomstart",()=>{
        followMoving=false;
        updateFollowButton();
    });

    map.on("click",async e=>{
        if(orderLocked || !mapPickMode) return;

        const point={
            lat:e.latlng.lat,
            lng:e.latlng.lng,
            name:"Vị trí chọn trên bản đồ"
        };

        const type=getMapClickTargetType();
        setPoint(type,point);
        await reverseName(point,type);

        mapPickMode=false;
        updateMapPickButton();

        followMoving=true;
        updateFollowButton();

        updateRouteAndWeather();
    });
}

function renderModeUI(){
    document.getElementById("ride-section").style.display=currentMode==="ride"?"block":"none";
    document.getElementById("food-section").style.display=currentMode==="food"?"block":"none";

    if(currentMode==="ride"){
        document.getElementById("mode-subtitle").innerText="Đặt xe thông minh";
        document.getElementById("price-label").innerText="DỰ KIẾN THANH TOÁN";
        document.getElementById("book-btn").innerText="ĐẶT XE NGAY";
        renderVehicles();
    }

    if(currentMode==="food"){
        document.getElementById("mode-subtitle").innerText="Đặt đồ ăn & nước uống";
        document.getElementById("price-label").innerText="TỔNG TIỀN ĐƠN HÀNG";
        document.getElementById("book-btn").innerText="ĐẶT ĐỒ ĂN";
        renderRestaurants();
        renderFoods();
        renderCart();
    }

    clearLiveMap();
    updateRouteAndWeather();
}

function renderVehicles(){
    const box=document.getElementById("vehicle-list");
    box.innerHTML="";

    Object.keys(vehicles).forEach(key=>{
        const v=vehicles[key];

        box.innerHTML+=`
            <div class="option-card ${key===selectedVehicle?"active":""}" onclick="selectVehicle('${key}')">
                <div class="option-top">
                    <div class="option-icon">${v.icon}</div>
                    <div>
                        <div class="option-name">${v.name}</div>
                        <div style="font-size:.78rem;color:#64748B;">Từ ${v.min.toLocaleString()}đ</div>
                    </div>
                </div>
                <div class="option-desc">${v.desc}</div>
            </div>
        `;
    });
}

function renderRestaurants(){
    const box=document.getElementById("restaurant-list");
    box.innerHTML="";

    Object.keys(restaurants).forEach(key=>{
        const r=restaurants[key];

        if(currentBrandFilter!=="all" && r.type!==currentBrandFilter){
            return;
        }

        box.innerHTML+=`
            <div class="food-brand-card ${key===selectedRestaurant?"active":""}" onclick="selectRestaurant('${key}')">
                <div class="brand-logo">${r.icon}</div>
                <div class="brand-name">${r.name}</div>
                <div class="brand-desc">${r.desc}</div>
                <div class="brand-meta">
                    <span class="brand-tag">⭐ ${r.rating}</span>
                    <span class="brand-tag">⏱️ ${r.time}</span>
                    <span class="brand-tag">Ship từ ${r.baseShip.toLocaleString()}đ</span>
                </div>
            </div>
        `;
    });
}

function filterFoodBrands(type,el){
    if(orderLocked) return;

    currentBrandFilter=type;

    document.querySelectorAll(".food-chip").forEach(chip=>chip.classList.remove("active"));
    el.classList.add("active");

    const available=Object.keys(restaurants).filter(key=>{
        return type==="all" || restaurants[key].type===type;
    });

    if(!available.includes(selectedRestaurant)){
        selectedRestaurant=available[0] || "kfc";
    }

    renderRestaurants();
    renderFoods();
    updateRouteAndWeather();
}

function renderFoods(){
    const inlineBox=document.getElementById("food-list");
    const modalBox=document.getElementById("food-modal-list");

    if(inlineBox){
        inlineBox.innerHTML=`
            <div style="background:white;border:1px solid #E2E8F0;border-radius:22px;padding:18px;text-align:center;">
                <div style="font-size:2.2rem;">${restaurants[selectedRestaurant].icon}</div>
                <h3 style="color:#0056B3;margin:8px 0;">${restaurants[selectedRestaurant].name}</h3>
                <p style="color:#64748B;font-weight:700;margin:0 0 14px 0;">
                    Bấm để xem thực đơn và thêm món vào giỏ hàng.
                </p>
                <button class="btn btn-primary" style="width:100%;" onclick="openFoodMenuModal()">
                    XEM THỰC ĐƠN
                </button>
            </div>
        `;
    }

    if(!modalBox) return;

    modalBox.innerHTML="";

    const menu=brandMenus[selectedRestaurant] || [];

    menu.forEach(group=>{
        modalBox.innerHTML+=`<div class="food-category-title">${group.category}</div>`;

        group.items.forEach(food=>{
            const icon=getFoodIcon(food.name);
            const basePrice=Math.min(food.prices.S,food.prices.M,food.prices.L);

            modalBox.innerHTML+=`
                <div class="food-card">
                    <div class="food-thumb">${icon}</div>
                    <div class="food-info">
                        <div class="food-price-line">
                            <h4>${food.name}</h4>
                            <div class="food-base-price">Từ ${basePrice.toLocaleString()}đ</div>
                        </div>

                        <small>${food.desc}</small>

                        <div class="size-row">
                            <button class="size-btn" onclick="addFood('${food.name}','S',${food.prices.S})">S<br>${food.prices.S.toLocaleString()}đ</button>
                            <button class="size-btn" onclick="addFood('${food.name}','M',${food.prices.M})">M<br>${food.prices.M.toLocaleString()}đ</button>
                            <button class="size-btn" onclick="addFood('${food.name}','L',${food.prices.L})">L<br>${food.prices.L.toLocaleString()}đ</button>
                        </div>
                    </div>
                </div>
            `;
        });
    });
}

function getFoodIcon(name){
    const lower=name.toLowerCase();

    for(const key in foodIcons){
        if(lower.includes(key)) return foodIcons[key];
    }

    return "🍽️";
}

function selectVehicle(key){
    if(orderLocked) return;
    selectedVehicle=key;
    renderVehicles();
    calculatePrice();
}

function selectRestaurant(key){
    if(orderLocked) return;

    selectedRestaurant=key;

    renderRestaurants();
    renderFoods();
    openFoodMenuModal();

    followMoving=true;
    updateFollowButton();

    updateRouteAndWeather();
}

function addFood(name,size,price){
    if(orderLocked) return;

    const brandKey=selectedRestaurant;
    const brandName=restaurants[selectedRestaurant].name;

    const found=cart.find(x=>x.name===name && x.size===size && x.brandKey===brandKey);

    if(found){
        found.qty+=1;
    }else{
        cart.push({
            brandKey,
            brandName,
            name,
            size,
            price,
            qty:1
        });
    }

    renderCart();
    calculatePrice();
}

function changeQty(index,delta){
    if(orderLocked) return;

    cart[index].qty+=delta;

    if(cart[index].qty<=0){
        cart.splice(index,1);
    }

    renderCart();
    calculatePrice();
}

function renderCart(){
    const box=document.getElementById("cart-items");
    const countBox=document.getElementById("cart-count");

    const itemCount=cart.reduce((sum,item)=>sum+item.qty,0);

    if(countBox){
        countBox.innerText=itemCount+" món";
    }

    if(cart.length===0){
        box.innerHTML=`<div class="cart-empty">Chưa có món nào. Hãy chọn món trong thực đơn phía trên.</div>`;
        return;
    }

    let foodTotal=0;
    let html="";
    const grouped={};

    cart.forEach(item=>{
        if(!grouped[item.brandName]){
            grouped[item.brandName]=[];
        }
        grouped[item.brandName].push(item);
    });

    Object.keys(grouped).forEach(brandName=>{
        html+=`<div style="margin-top:12px;font-weight:1000;color:#0056B3;">🏪 ${brandName}</div>`;

        grouped[brandName].forEach(item=>{
            const index=cart.indexOf(item);
            const sub=item.price*item.qty;
            foodTotal+=sub;

            html+=`
                <div class="cart-row">
                    <div>
                        <div class="cart-item-name">${item.name}</div>
                        <div class="cart-item-meta">Size ${item.size} • ${item.price.toLocaleString()}đ x ${item.qty}</div>
                    </div>
                    <div class="cart-controls">
                        <button class="qty-btn minus" onclick="changeQty(${index},-1)">-</button>
                        <b>${item.qty}</b>
                        <button class="qty-btn plus" onclick="changeQty(${index},1)">+</button>
                    </div>
                </div>
            `;
        });
    });

    html+=`
        <div class="cart-total-box">
            <div class="cart-total-row">
                <span>Tạm tính món ăn</span>
                <span>${foodTotal.toLocaleString()}đ</span>
            </div>
            <div class="cart-total-row">
                <span>Phí giao hàng</span>
                <span>Tính theo bản đồ</span>
            </div>
            <div class="cart-total-row main">
                <span>Tổng đơn dự kiến</span>
                <span>${finalPrice.toLocaleString()}đ</span>
            </div>
        </div>
    `;

    box.innerHTML=html;
}

function getFoodTotal(){
    return cart.reduce((sum,item)=>sum+item.price*item.qty,0);
}

function openFoodMenuModal(){
    if(currentMode!=="food") return;

    const r=restaurants[selectedRestaurant];

    document.getElementById("food-modal-title").innerText=r.icon+" "+r.name;
    document.getElementById("food-modal-sub").innerText=r.desc+" • ⭐ "+r.rating+" • "+r.time;

    renderFoods();

    document.getElementById("food-menu-modal").style.display="flex";
}

function closeFoodMenuModal(){
    document.getElementById("food-menu-modal").style.display="none";
}

function selectPromo(code,el){
    if(orderLocked) return;

    document.querySelectorAll(".promo-card").forEach(card=>card.classList.remove("active"));

    if(selectedPromo===code){
        selectedPromo="";
    }else{
        selectedPromo=code;
        el.classList.add("active");
    }

    calculatePrice();
}

function loadDefaultPositionForMode(){
    if(currentMode==="ride"){
        ridePickup={...defaultPoints.ridePickup};
        rideDest={...defaultPoints.rideDest};
        setInputValue("ridePickup",ridePickup.name);
        setInputValue("rideDest",rideDest.name);
    }

    if(currentMode==="food"){
        foodDrop=foodDrop || {...defaultPoints.foodDrop};
        setInputValue("foodDrop",foodDrop.name);
    }

    updateRouteAndWeather();
}

function getInputId(type){
    return {
        ridePickup:"ride-pickup-input",
        rideDest:"ride-dest-input",
        foodDrop:"food-drop-input"
    }[type];
}

function setInputValue(type,value){
    const el=document.getElementById(getInputId(type));
    if(el) el.value=value;
}

function setPoint(type,point){
    if(type==="ridePickup") ridePickup=point;
    if(type==="rideDest") rideDest=point;
    if(type==="foodDrop") foodDrop=point;

    setInputValue(type,point.name);
}

function getMapClickTargetType(){
    if(currentMode==="ride"){
        return ridePickup ? "rideDest" : "ridePickup";
    }

    return "foodDrop";
}

function searchPlaceDebounced(input,type){
    if(orderLocked) return;

    clearTimeout(debounceTimer);

    debounceTimer=setTimeout(()=>{
        searchPlace(input,type);
    },450);
}

async function searchPlace(input,type){
    if(orderLocked) return;

    const q=input.value.trim();
    const box=input.nextElementSibling;

    if(q.length<2){
        box.style.display="none";
        return;
    }

    box.innerHTML=`<div class="suggest-item"><div class="suggest-name">Đang tìm...</div></div>`;
    box.style.display="block";

    try{
        const params=new URLSearchParams({
            q,
            format:"jsonv2",
            addressdetails:"1",
            limit:"7",
            countrycodes:"vn",
            accept_language:"vi"
        });

        const res=await fetch(`${NOMINATIM}/search?${params.toString()}`);
        const data=await res.json();

        if(!data.length){
            box.innerHTML=`<div class="suggest-item"><div class="suggest-name">Không tìm thấy địa điểm</div></div>`;
            return;
        }

        box.innerHTML=data.map(item=>{
            const safeName=escapeHtml(item.display_name || "Địa điểm");

            return `
                <div class="suggest-item" onclick="choosePlace('${type}',${item.lat},${item.lon},'${encodeURIComponent(safeName)}',this)">
                    <div class="suggest-name">📍 ${safeName.split(",")[0]}</div>
                    <div class="suggest-sub">${safeName}</div>
                </div>
            `;
        }).join("");
    }catch(err){
        box.innerHTML=`<div class="suggest-item"><div class="suggest-name">Lỗi gọi API tìm kiếm</div></div>`;
        console.log(err);
    }
}

async function choosePlace(type,lat,lng,encodedName,el){
    if(orderLocked) return;

    const point={
        lat:parseFloat(lat),
        lng:parseFloat(lng),
        name:decodeURIComponent(encodedName)
    };

    setPoint(type,point);

    el.closest(".suggest-box").style.display="none";

    followMoving=true;
    updateFollowButton();

    updateRouteAndWeather();
}

async function reverseName(point,type){
    try{
        const url=`${NOMINATIM}/reverse?format=jsonv2&lat=${point.lat}&lon=${point.lng}&accept-language=vi`;
        const res=await fetch(url);
        const data=await res.json();

        point.name=data.display_name || point.name;
        setInputValue(type,point.name);
    }catch(err){
        console.log(err);
    }
}

function getCurrentStartEnd(){
    if(currentMode==="ride"){
        return {start:ridePickup,end:rideDest};
    }

    if(currentMode==="food"){
        const r=restaurants[selectedRestaurant];
        return {
            start:{lat:r.lat,lng:r.lng,name:r.name},
            end:foodDrop
        };
    }

    return {start:null,end:null};
}

async function updateRouteAndWeather(){
    calculatePeakAndTraffic();

    const {start,end}=getCurrentStartEnd();

    if(start){
        await loadWeather(start.lat,start.lng);
    }

    updateMarkers();

    if(start && end){
        await getRoute(start,end);
    }else{
        routeDistanceKm=0;
        routeDurationMin=0;
        document.getElementById("route-status").innerText="Chọn đủ điểm để tính tuyến đường";
    }

    calculatePrice();
}

function updateMarkers(){
    if(!map) return;

    if(pickupMarker) map.removeLayer(pickupMarker);
    if(destMarker) map.removeLayer(destMarker);

    const {start,end}=getCurrentStartEnd();
    const startRole=getStartRole();
    const endRole=getEndRole();

    if(start){
        pickupMarker=createPointMarker(start,startRole);
    }

    if(end){
        destMarker=createPointMarker(end,endRole);
    }

    if(start && end){
        smartFitBounds(L.latLngBounds([[start.lat,start.lng],[end.lat,end.lng]]));
    }else if(start && followMoving){
        map.setView([start.lat,start.lng],15);
    }
}

function getStartRole(){
    if(currentMode==="ride") return "pickup";
    if(currentMode==="food") return "restaurant";
    return "pickup";
}

function getEndRole(){
    if(currentMode==="ride") return "dropoff";
    if(currentMode==="food") return "delivery";
    return "dropoff";
}

function getPointVisual(role){
    const data={
        pickup:{emoji:"📍",label:"Điểm đón",popup:"Điểm đón"},
        dropoff:{emoji:"🏁",label:"Trả khách",popup:"Điểm trả khách"},
        restaurant:{emoji:"🏪",label:"Nhà hàng",popup:"Điểm lấy hàng"},
        delivery:{emoji:"📦",label:"Nhận hàng",popup:"Điểm nhận hàng"}
    };

    return data[role] || data.pickup;
}

function createPointMarker(point,role){
    const visual=getPointVisual(role);

    const icon=L.divIcon({
        className:"",
        html:`<div class="point-marker ${role}">
                <div class="pin">${visual.emoji} ${visual.label}</div>
              </div>`,
        iconSize:[120,44],
        iconAnchor:[60,44],
        popupAnchor:[0,-44]
    });

    return L.marker([point.lat,point.lng],{icon})
        .addTo(map)
        .bindPopup(`<b>${visual.popup}</b><br>${point.name}`);
}

async function getRoute(start,end){
    document.getElementById("route-status").innerText="Đang tính tuyến đường thật...";

    try{
        const coords=`${start.lng},${start.lat};${end.lng},${end.lat}`;
        const url=`${OSRM}/route/v1/driving/${coords}?overview=full&geometries=geojson&steps=true`;

        const res=await fetch(url);
        const data=await res.json();

        if(!data.routes || !data.routes.length){
            throw new Error("Không tìm thấy route");
        }

        const route=data.routes[0];
        routeDistanceKm=route.distance/1000;
        routeDurationMin=Math.ceil(route.duration/60);

        const latlngs=route.geometry.coordinates.map(c=>[c[1],c[0]]);

        if(routeLayer) map.removeLayer(routeLayer);
        if(trafficLayer) map.removeLayer(trafficLayer);

        routeLayer=L.polyline(latlngs,{
            color:currentMode==="food"?"#FF6B00":"#0056B3",
            weight:7,
            opacity:.9
        }).addTo(map);

        trafficLayer=L.polyline(latlngs,{
            color:getTrafficColor(),
            weight:11,
            opacity:.32
        }).addTo(map);

        smartFitBounds(routeLayer.getBounds());

        document.getElementById("route-status").innerText=`${routeDistanceKm.toFixed(2)} km • ${routeDurationMin} phút`;

        drawPeakArea();

    }catch(err){
        document.getElementById("route-status").innerText="Không tính được tuyến đường";
        console.log(err);
    }
}

async function loadWeather(lat,lng){
    try{
        const params=new URLSearchParams({
            latitude:lat,
            longitude:lng,
            current:"temperature_2m,precipitation,rain,weather_code,wind_speed_10m",
            timezone:"auto"
        });

        const res=await fetch(`${OPEN_METEO}?${params.toString()}`);
        const data=await res.json();

        applyWeather(data.current);

    }catch(err){
        weatherMultiplier=1;
        document.getElementById("weather-title").innerText="Không lấy được thời tiết";
        document.getElementById("weather-detail").innerText="Dùng hệ số x1.00";
    }
}

function applyWeather(w){
    if(!w){
        weatherMultiplier=1;
        return;
    }

    const code=w.weather_code;
    const temp=Math.round(w.temperature_2m);
    const rain=w.rain || w.precipitation || 0;
    const wind=Math.round(w.wind_speed_10m || 0);

    let text="Trời đẹp";
    let icon="🌤️";
    weatherMultiplier=1;

    if(code>=95){
        text="Dông, mưa mạnh";
        icon="⛈️";
        weatherMultiplier=1.35;
    }else if(code>=80 || rain>=2){
        text="Mưa rào";
        icon="🌧️";
        weatherMultiplier=1.25;
    }else if(code>=61 || rain>0){
        text="Mưa nhẹ";
        icon="🌦️";
        weatherMultiplier=1.15;
    }else if(code>=45){
        text="Sương mù";
        icon="🌫️";
        weatherMultiplier=1.1;
    }else if(temp>=35){
        text="Nắng nóng";
        icon="☀️";
        weatherMultiplier=1.08;
    }

    document.getElementById("weather-icon").innerText=icon;
    document.getElementById("weather-title").innerText=text;
    document.getElementById("weather-detail").innerText=`${temp}°C • mưa ${rain} mm • gió ${wind} km/h • hệ số x${weatherMultiplier.toFixed(2)}`;
    document.getElementById("weather-meta").innerText=`${text} x${weatherMultiplier.toFixed(2)}`;
}

function calculatePeakAndTraffic(){
    const now=new Date();
    const h=now.getHours();
    const m=now.getMinutes();

    let peakText="Không cao điểm";
    peakMultiplier=1;
    trafficMultiplier=1;

    if((h>=7 && h<=9) || (h>=17 && h<=19)){
        peakText="Đang giờ cao điểm";
        peakMultiplier=1.28;
        trafficMultiplier=1.25;
    }else if((h>=11 && h<=12) || (h>=21 && h<=22)){
        peakText="Mật độ trung bình";
        peakMultiplier=1.1;
        trafficMultiplier=1.12;
    }else if(h>=0 && h<=5){
        peakText="Đêm khuya";
        peakMultiplier=1.12;
        trafficMultiplier=.92;
    }

    document.getElementById("peak-map-status").innerText=`${peakText} lúc ${String(h).padStart(2,"0")}:${String(m).padStart(2,"0")} • x${peakMultiplier.toFixed(2)}`;
    document.getElementById("traffic-map-status").innerText=`Hệ số tuyến hiện tại x${trafficMultiplier.toFixed(2)}`;
    document.getElementById("peak-meta").innerText=`Cao điểm x${peakMultiplier.toFixed(2)}`;
}

function drawPeakArea(){
    if(peakCircle) map.removeLayer(peakCircle);

    const {start}=getCurrentStartEnd();

    if(!start) return;

    const color=peakMultiplier>1.2?"#FF6B00":"#22C55E";

    peakCircle=L.circle([start.lat,start.lng],{
        radius:1200,
        color,
        fillColor:color,
        fillOpacity:.12,
        weight:2
    }).addTo(map);

    peakCircle.bindPopup(`
        <b>Điều kiện khu vực</b><br>
        Giờ cao điểm: x${peakMultiplier.toFixed(2)}<br>
        Mật độ tuyến: x${trafficMultiplier.toFixed(2)}
    `);
}

function getTrafficColor(){
    if(trafficMultiplier>=1.2) return "#FF6B00";
    if(trafficMultiplier>=1.1) return "#F59E0B";
    return "#22C55E";
}

function calculatePrice(){
    let rows="";

    if(currentMode==="ride"){
        const v=vehicles[selectedVehicle];
        const distanceFee=routeDistanceKm*v.perKm;

        let subtotal=v.base+distanceFee+v.platform;

        if(subtotal<v.min) subtotal=v.min;

        const dynamic=subtotal*weatherMultiplier*peakMultiplier*trafficMultiplier;
        const discount=getDiscount(dynamic);
        finalPrice=Math.max(0,Math.round((dynamic-discount)/1000)*1000);

        const adjustedDuration=Math.ceil(routeDurationMin*trafficMultiplier/v.speedFactor);

        rows=`
            <div class="break-row"><span>Dịch vụ</span><b>Đặt xe ${v.icon} ${v.name}</b></div>
            <div class="break-row"><span>Quãng đường</span><b>${routeDistanceKm.toFixed(2)} km</b></div>
            <div class="break-row"><span>Thời gian</span><b>${adjustedDuration || 0} phút</b></div>
            <div class="break-row"><span>Giá mở cửa</span><b>${v.base.toLocaleString()}đ</b></div>
            <div class="break-row"><span>Giá theo km</span><b>${Math.round(distanceFee).toLocaleString()}đ</b></div>
            <div class="break-row"><span>Phí nền tảng</span><b>${v.platform.toLocaleString()}đ</b></div>
            <div class="break-row"><span>Thời tiết</span><b>x${weatherMultiplier.toFixed(2)}</b></div>
            <div class="break-row"><span>Cao điểm</span><b>x${peakMultiplier.toFixed(2)}</b></div>
            <div class="break-row"><span>Mật độ tuyến</span><b>x${trafficMultiplier.toFixed(2)}</b></div>
            <div class="break-row"><span>Giảm giá</span><b>-${Math.round(discount).toLocaleString()}đ</b></div>
            <div class="break-row"><span>Thành tiền</span><b>${finalPrice.toLocaleString()}đ</b></div>
        `;

        document.getElementById("duration-meta").innerText=`${adjustedDuration || 0} phút`;
    }

    if(currentMode==="food"){
        const r=restaurants[selectedRestaurant];
        const foodTotal=getFoodTotal();
        const shipRaw=r.baseShip+routeDistanceKm*r.perKm;
        const shipFee=shipRaw*weatherMultiplier*peakMultiplier*trafficMultiplier;
        const serviceFee=3000;

        const subtotal=foodTotal+shipFee+serviceFee;
        const discount=getDiscount(subtotal);

        finalPrice=Math.max(0,Math.round((subtotal-discount)/1000)*1000);

        const adjustedDuration=Math.ceil(routeDurationMin*trafficMultiplier+12);

        rows=`
            <div class="break-row"><span>Dịch vụ</span><b>Đặt đồ ăn 🍔</b></div>
            <div class="break-row"><span>Thương hiệu hiện tại</span><b>${r.name}</b></div>
            <div class="break-row"><span>Tiền món</span><b>${foodTotal.toLocaleString()}đ</b></div>
            <div class="break-row"><span>Quãng đường giao</span><b>${routeDistanceKm.toFixed(2)} km</b></div>
            <div class="break-row"><span>Thời gian giao</span><b>${adjustedDuration || 0} phút</b></div>
            <div class="break-row"><span>Phí ship nền</span><b>${Math.round(shipRaw).toLocaleString()}đ</b></div>
            <div class="break-row"><span>Thời tiết</span><b>x${weatherMultiplier.toFixed(2)}</b></div>
            <div class="break-row"><span>Cao điểm</span><b>x${peakMultiplier.toFixed(2)}</b></div>
            <div class="break-row"><span>Mật độ tuyến</span><b>x${trafficMultiplier.toFixed(2)}</b></div>
            <div class="break-row"><span>Phí dịch vụ</span><b>${serviceFee.toLocaleString()}đ</b></div>
            <div class="break-row"><span>Giảm giá</span><b>-${Math.round(discount).toLocaleString()}đ</b></div>
            <div class="break-row"><span>Thành tiền</span><b>${finalPrice.toLocaleString()}đ</b></div>
        `;

        document.getElementById("duration-meta").innerText=`${adjustedDuration || 0} phút`;
    }

    document.getElementById("total-price").innerText=finalPrice.toLocaleString()+" VNĐ";
    document.getElementById("distance-meta").innerText=routeDistanceKm.toFixed(2)+" km";
    document.getElementById("breakdown").innerHTML=rows;
    document.getElementById("promo-note").innerText=selectedPromo ? "Đã chọn mã: "+selectedPromo : "";

    if(currentMode==="food"){
        renderCart();
    }
}

function getDiscount(amount){
    if(selectedPromo==="GO20") return Math.min(amount*.2,30000);
    if(selectedPromo==="NEWUSER") return Math.min(amount*.35,50000);
    if(selectedPromo==="STUDENT") return Math.min(amount*.15,20000);
    return 0;
}

function confirmService(){
    if(orderLocked) return;

    const {start,end}=getCurrentStartEnd();

    if(!start || !end || routeDistanceKm<=0){
        alert("Vui lòng chọn đủ vị trí hợp lệ.");
        return;
    }

    if(currentMode==="food" && cart.length===0){
        alert("Vui lòng chọn ít nhất một món ăn.");
        return;
    }

    lockUI();

    if(currentMode==="ride"){
        openModal("Đang tìm tài xế...","Hệ thống đang tìm tài xế gần bạn.");
        setTimeout(()=>startRideFlow(),1700);
    }

    if(currentMode==="food"){
        openModal("Đang tìm shipper...","Nhà hàng đang xác nhận đơn và tìm shipper.");
        setTimeout(()=>startFoodFlow(),1700);
    }
}

function startRideFlow(){
    const driver=drivers[Math.floor(Math.random()*drivers.length)];

    setupLiveCard({
        avatar:driver.icon,
        name:driver.name,
        vehicle:`${driver.vehicle} • ${driver.plate}`,
        rating:"⭐ 4.9 • 1.248 chuyến",
        code:"TG-"+randomCode(),
        stage1:"Nhận chuyến",
        stage2:"Đón khách",
        stage3:"Di chuyển",
        stage4:"Hoàn thành"
    });

    closeModalAfterSuccess("Đã tìm thấy tài xế","Tài xế đang đến điểm đón.");

    animateApproachToStart("Tài xế đang đến điểm đón","Tài xế đã đến điểm đón",()=>{
        setTimeout(()=>animateMainRoute("Đang chở khách đến điểm đến","Đã trả khách thành công"),1600);
    });
}

function startFoodFlow(){
    const shipper=shippers[Math.floor(Math.random()*shippers.length)];

    setupLiveCard({
        avatar:shipper.icon,
        name:shipper.name,
        vehicle:`${shipper.vehicle} • ${shipper.plate}`,
        rating:"⭐ 4.8 • Giao nhanh",
        code:"FD-"+randomCode(),
        stage1:"Nhận đơn",
        stage2:"Lấy món",
        stage3:"Đang giao",
        stage4:"Hoàn thành"
    });

    closeModalAfterSuccess("Shipper đã nhận đơn","Shipper đang đến nhà hàng lấy món.");

    animateApproachToStart("Shipper đang đến nhà hàng","Shipper đã lấy món",()=>{
        setTimeout(()=>animateMainRoute("Shipper đang giao đồ ăn","Đã giao đồ ăn thành công"),1600);
    });
}

function setupLiveCard(data){
    document.getElementById("live-avatar").innerText=data.avatar;
    document.getElementById("live-name").innerText=data.name;
    document.getElementById("live-vehicle").innerText=data.vehicle;
    document.getElementById("live-rating").innerText=data.rating;
    document.getElementById("order-code").innerText=data.code;

    document.getElementById("stage-1").innerText=data.stage1;
    document.getElementById("stage-2").innerText=data.stage2;
    document.getElementById("stage-3").innerText=data.stage3;
    document.getElementById("stage-4").innerText=data.stage4;

    document.getElementById("live-card").style.display="block";

    setStage(1,15,"Đối tác đã nhận đơn");
}

async function animateApproachToStart(runningText,doneText,callback){
    const {start}=getCurrentStartEnd();

    if(!start) return;

    const fakeStart={
        lat:start.lat+0.018,
        lng:start.lng-0.02,
        name:"Đối tác gần bạn"
    };

    if(movingMarker) map.removeLayer(movingMarker);
    if(approachLayer) map.removeLayer(approachLayer);

    movingMarker=L.marker([fakeStart.lat,fakeStart.lng],{
        icon:getMovingVehicleIcon()
    }).addTo(map).bindPopup("Đối tác của bạn");

    try{
        const coords=`${fakeStart.lng},${fakeStart.lat};${start.lng},${start.lat}`;
        const res=await fetch(`${OSRM}/route/v1/driving/${coords}?overview=full&geometries=geojson`);
        const data=await res.json();

        const route=data.routes[0];
        const latlngs=route.geometry.coordinates.map(c=>[c[1],c[0]]);
        const eta=Math.max(2,Math.ceil(route.duration/60));

        approachLayer=L.polyline(latlngs,{
            color:"#FF6B00",
            weight:5,
            opacity:.8,
            dashArray:"8 8"
        }).addTo(map);

        setStage(2,40,runningText);

        animateMarker(latlngs,eta,start,()=>{
            setStage(2,50,doneText);
            if(callback) callback();
        });

    }catch(err){
        console.log(err);
    }
}

function animateMainRoute(runningText,doneText){
    if(!routeLayer || !movingMarker) return;

    const latlngs=routeLayer.getLatLngs();

    setStage(3,70,runningText);

    animateMarker(latlngs,routeDurationMin,getCurrentStartEnd().end,()=>{
        setStage(4,100,doneText);
        document.getElementById("system-status").innerText="● Hoàn thành";

        setTimeout(()=>{
            showRatingModal();
        },500);
    },70);
}

function animateMarker(latlngs,minutes,targetPoint,callback,startProgress=40){
    if(liveTimer) clearInterval(liveTimer);

    let i=0;
    const total=latlngs.length;
    const interval=145;

    liveTimer=setInterval(()=>{
        if(i>=total){
            clearInterval(liveTimer);
            if(callback) callback();
            return;
        }

        const current=latlngs[i];

        movingMarker.setLatLng(current);
        smartPanTo(current);

        const ratio=i/total;
        const remain=Math.max(0,Math.ceil(minutes*(1-ratio)));

        document.getElementById("live-eta").innerText=remain===0 ? "Sắp đến" : remain+" phút";

        const currentPoint={lat:current[0],lng:current[1]};
        const dist=estimateDistanceBetween(currentPoint,targetPoint);

        document.getElementById("live-distance").innerText=dist.toFixed(2)+" km";

        if(startProgress>=70){
            document.getElementById("trip-progress-fill").style.width=(70+ratio*30)+"%";
        }

        i++;
    },interval);
}

function setStage(stage,progress,text){
    ["stage-1","stage-2","stage-3","stage-4"].forEach(id=>{
        document.getElementById(id).classList.remove("active");
    });

    document.getElementById("stage-"+stage).classList.add("active");
    document.getElementById("trip-progress-fill").style.width=progress+"%";
    document.getElementById("live-trip-status").innerText=text;
    document.getElementById("system-status").innerText="● "+text;
}

function getMovingVehicleEmoji(){
    if(currentMode==="ride"){
        return selectedVehicle==="bike" ? "🏍️" : selectedVehicle==="xl" ? "🚐" : "🚗";
    }

    return "🛵";
}

function getMovingVehicleIcon(){
    return L.divIcon({
        className:"",
        html:`<div class="moving-vehicle-marker ${currentMode}">${getMovingVehicleEmoji()}</div>`,
        iconSize:[48,48],
        iconAnchor:[24,24],
        popupAnchor:[0,-26]
    });
}

function toggleMapPickMode(){
    if(orderLocked) return;

    mapPickMode=!mapPickMode;
    updateMapPickButton();

    if(mapPickMode){
        alert("Bấm 1 điểm trên bản đồ để chọn vị trí. Sau khi chọn xong, chế độ này sẽ tự tắt.");
    }
}

function updateMapPickButton(){
    const btn=document.getElementById("map-pick-btn");

    if(!btn) return;

    if(mapPickMode){
        btn.classList.add("active");
        btn.innerText="📍 Chọn trên bản đồ: Bật";
    }else{
        btn.classList.remove("active");
        btn.innerText="📍 Chọn trên bản đồ: Tắt";
    }
}

function toggleFollowMode(){
    followMoving=!followMoving;
    updateFollowButton();
}

function updateFollowButton(){
    const btn=document.getElementById("follow-btn");

    if(!btn) return;

    if(followMoving){
        btn.classList.add("active");
        btn.innerText="🎯 Tự theo dõi: Bật";
    }else{
        btn.classList.remove("active");
        btn.innerText="🎯 Tự theo dõi: Tắt";
    }
}

function focusCurrentRoute(){
    followMoving=true;
    updateFollowButton();

    if(routeLayer){
        map.fitBounds(routeLayer.getBounds(),{padding:[90,90]});
        return;
    }

    const {start,end}=getCurrentStartEnd();

    if(start && end){
        map.fitBounds(L.latLngBounds([[start.lat,start.lng],[end.lat,end.lng]]),{padding:[90,90]});
    }else if(start){
        map.setView([start.lat,start.lng],15);
    }
}

function smartPanTo(latlng){
    if(followMoving){
        map.panTo(latlng,{animate:true});
    }
}

function smartFitBounds(bounds){
    if(followMoving){
        map.fitBounds(bounds,{padding:[90,90]});
    }
}

function openModal(title,text){
    document.getElementById("process-modal").style.display="flex";
    document.getElementById("modal-loader").style.display="block";
    document.getElementById("modal-title").innerText=title;
    document.getElementById("modal-text").innerText=text;
    document.getElementById("modal-close").style.display="none";
}

function closeModalAfterSuccess(title,text){
    document.getElementById("modal-loader").style.display="none";
    document.getElementById("modal-title").innerText=title;
    document.getElementById("modal-text").innerText=text;
    document.getElementById("modal-close").style.display="block";
}

function closeModal(){
    document.getElementById("process-modal").style.display="none";
}

function lockUI(){
    orderLocked=true;

    document.querySelectorAll("input,select,button").forEach(el=>{
        if(el.closest("#rating-modal")) return;

        if(
            el.id!=="modal-close" &&
            !el.classList.contains("btn-danger") &&
            !el.classList.contains("small-btn") &&
            !el.classList.contains("close-food-btn")
        ){
            el.disabled=true;
        }
    });

    document.getElementById("promo-grid").classList.add("locked-overlay");
    document.getElementById("system-status").innerText="● Đang xử lý đơn";
}

function unlockUI(){
    orderLocked=false;

    document.querySelectorAll("input,select,button").forEach(el=>{
        el.disabled=false;
    });

    document.getElementById("promo-grid").classList.remove("locked-overlay");
}

function cancelLiveOrder(){
    if(liveTimer) clearInterval(liveTimer);

    if(movingMarker){
        map.removeLayer(movingMarker);
        movingMarker=null;
    }

    if(approachLayer){
        map.removeLayer(approachLayer);
        approachLayer=null;
    }

    document.getElementById("live-card").style.display="none";
    document.getElementById("system-status").innerText="● Đã hủy";

    unlockUI();
}

function resetCurrentService(){
    if(orderLocked){
        alert("Bạn đang có đơn/chuyến đang chạy. Hãy hủy hoặc chờ hoàn thành.");
        return;
    }

    clearLiveMap();

    if(currentMode==="ride"){
        ridePickup=null;
        rideDest=null;
        selectedVehicle="bike";
        document.getElementById("ride-pickup-input").value="";
        document.getElementById("ride-dest-input").value="";
        renderVehicles();
    }

    if(currentMode==="food"){
        foodDrop=null;
        cart=[];
        selectedRestaurant="kfc";
        currentBrandFilter="all";
        document.getElementById("food-drop-input").value="";

        document.querySelectorAll(".food-chip").forEach(chip=>chip.classList.remove("active"));
        const firstChip=document.querySelector(".food-chip");
        if(firstChip) firstChip.classList.add("active");

        renderRestaurants();
        renderFoods();
        renderCart();
    }

    selectedPromo="";
    document.querySelectorAll(".promo-card").forEach(card=>card.classList.remove("active"));

    routeDistanceKm=0;
    routeDurationMin=0;
    finalPrice=0;

    document.getElementById("total-price").innerText="0 VNĐ";
    document.getElementById("distance-meta").innerText="0 km";
    document.getElementById("duration-meta").innerText="0 phút";
    document.getElementById("breakdown").innerHTML="";
    document.getElementById("promo-note").innerText="";
    document.getElementById("route-status").innerText="Chọn vị trí để tính tuyến đường";

    followMoving=true;
    updateFollowButton();

    loadDefaultPositionForMode();
}

function clearLiveMap(){
    [pickupMarker,destMarker,routeLayer,trafficLayer,peakCircle,movingMarker,approachLayer].forEach(layer=>{
        if(layer) map.removeLayer(layer);
    });

    pickupMarker=null;
    destMarker=null;
    routeLayer=null;
    trafficLayer=null;
    peakCircle=null;
    movingMarker=null;
    approachLayer=null;

    document.getElementById("live-card").style.display="none";
    document.getElementById("trip-progress-fill").style.width="0%";
}

function showRatingModal(){
    const modal=document.getElementById("rating-modal");
    const title=document.getElementById("rating-title");
    const sub=document.getElementById("rating-sub");
    const box=document.getElementById("rating-criteria");

    ratingData={};

    if(currentMode==="ride"){
        title.innerText="Đánh giá chuyến xe";
        sub.innerText="Bạn thấy tài xế và chuyến đi như thế nào?";
    }

    if(currentMode==="food"){
        title.innerText="Đánh giá đơn đồ ăn";
        sub.innerText="Đánh giá shipper và thương hiệu bạn vừa đặt.";
    }

    const criteria=ratingCriteriaByMode[currentMode];

    box.innerHTML="";

    criteria.forEach((name,index)=>{
        ratingData[name]=5;

        box.innerHTML+=`
            <div class="rating-row">
                <div class="rating-label">${name}</div>
                <div class="stars" id="stars-${index}">
                    ${[1,2,3,4,5].map(star=>`
                        <span class="star active" onclick="setRating('${name}', ${star}, ${index})">★</span>
                    `).join("")}
                </div>
            </div>
        `;
    });

    modal.style.display="flex";

    const submitBtn=document.querySelector(".rating-submit-btn");
    if(submitBtn) submitBtn.disabled=false;

    const noteBox=document.getElementById("rating-note");
    if(noteBox) noteBox.disabled=false;
}

function setRating(name,value,index){
    ratingData[name]=value;

    document.querySelectorAll(`#stars-${index} .star`).forEach((star,i)=>{
        if(i<value){
            star.classList.add("active");
        }else{
            star.classList.remove("active");
        }
    });
}

function submitRating(){
    const note=document.getElementById("rating-note").value.trim();

    console.log("Đánh giá:",{
        mode:currentMode,
        rating:ratingData,
        note
    });

    document.getElementById("rating-modal").style.display="none";
    document.getElementById("rating-note").value="";

    alert("Cảm ơn bạn đã đánh giá dịch vụ!");

    unlockUI();
}

function fakeCall(){
    alert("Đang gọi đối tác...");
}

function fakeChat(){
    alert("Mở khung chat với đối tác...");
}

function randomCode(){
    return Math.floor(1000+Math.random()*9000);
}

function estimateDistanceBetween(a,b){
    if(!a || !b) return 0;

    const R=6371;
    const dLat=(b.lat-a.lat)*Math.PI/180;
    const dLng=(b.lng-a.lng)*Math.PI/180;
    const lat1=a.lat*Math.PI/180;
    const lat2=b.lat*Math.PI/180;

    const x=
        Math.sin(dLat/2)**2+
        Math.cos(lat1)*Math.cos(lat2)*Math.sin(dLng/2)**2;

    return R*2*Math.atan2(Math.sqrt(x),Math.sqrt(1-x));
}

function escapeHtml(text){
    return String(text)
        .replaceAll("&","&amp;")
        .replaceAll("<","&lt;")
        .replaceAll(">","&gt;")
        .replaceAll('"',"&quot;")
        .replaceAll("'","&#039;");
}
</script>
</body>
</html>
'''

display(HTML(html_code))



```
